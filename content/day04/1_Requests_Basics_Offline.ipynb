## <p style="text-align: right; direction: rtl; float: right; clear: both;">הבקשה הראשונה שלנו!</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בואו נשלח את הבקשה הראשונה שלנו - נבקש מידע על מזג האוויר בתל אביב.<br>
שימו לב לשתי תוספות חשובות: <code>timeout=10</code> (מחכים לתשובה לכל היותר 10 שניות)
ו-<code>try / except</code> (אם משהו משתבש, התוכנית לא קורסת אלא מדפיסה הודעה).
</p>

```python
import requests

try:
    # שולחים בקשה לקבלת מזג אוויר. timeout=10 = לא מחכים יותר מ-10 שניות
    response = requests.get('https://wttr.in/TelAviv?format=3', timeout=10)

    # מה קיבלנו בחזרה?
    print("מזג האוויר בתל אביב:")
    print(response.text)
except requests.exceptions.RequestException:
    print("⚠️ לא הצלחנו להתחבר לשירות כרגע. זה קורה - ממשיכים לדוגמה הבאה!")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 <code>requests.get()</code> - שלחנו בקשת GET ("תן לי מידע")<br>
🔹 העברנו כתובת של שירות מזג אוויר<br>
🔹 <code>response.text</code> - קיבלנו את התשובה כטקסט<br>
🔹 <code>timeout=10</code> - בלי זה, אם השרת לא עונה, התוכנית עלולה לתקוע אתכם לנצח<br>
🔹 <code>try / except</code> - אם אין אינטרנט או שהשרת עמוס, מקבלים הודעה ידידותית במקום שגיאה אדומה<br><br>
💡 <strong>חשוב:</strong> wttr.in הוא שירות חינמי שכל הכיתה פונה אליו מאותה כתובת רשת,
ולכן לפעמים הוא יענה לאט או יסרב. זה לא באג בקוד שלכם! בהמשך נכיר שירות אמין יותר (Open-Meteo).
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">פונקציית עזר: safe_get</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
כדי לא לכתוב <code>timeout</code> ו-<code>try / except</code> בכל בקשה מחדש, נכתוב פעם אחת פונקציה קטנה.<br>
היא שולחת בקשה עם timeout של 10 שניות, ואם יש תקלה - מדפיסה הודעה ומחזירה <code>None</code> (כלום).<br>
את הפונקציה הזו נשתמש בכל שאר המחברת.
</p>

```python
def safe_get(url, **kwargs):
    """שולחת בקשת GET עם timeout. אם יש תקלה - מחזירה None."""
    try:
        return requests.get(url, timeout=10, **kwargs)
    except requests.exceptions.RequestException:
        print("⚠️ לא הצלחנו להתחבר לשירות כרגע (אין אינטרנט או שהשרת עמוס).")
        return None


print("הפונקציה מוכנה! ✅")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 <code>**kwargs</code> מאפשר להעביר לפונקציה כל פרמטר נוסף של <code>requests.get</code> (כמו <code>params</code> או <code>headers</code>)<br>
🔹 כשיש תקלה מקבלים <code>None</code>, ולכן אחרי כל בקשה בודקים <code>if response is not None</code><br><br>
⚠️ שימו לב: כדאי לכתוב <code>response is not None</code> ולא <code>if response:</code> -
ב-requests תשובה עם קוד שגיאה (כמו 404) נחשבת "שקרית", ואנחנו רוצים להבדיל בין "אין תשובה בכלל" לבין "יש תשובה עם שגיאה".
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">בדיקת סטטוס הבקשה</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
לא תמיד הבקשה מצליחה. חשוב לבדוק שהכל עבד כמו שצריך:<br>
🔹 קוד 200 = הכל בסדר ✅<br>
🔹 קוד 404 = לא מצאו את מה שביקשת ❌<br>
🔹 קוד 500 = שגיאה בשרת 🔥<br>
</p>

```python
response = safe_get('https://wttr.in/Jerusalem?format=3')

if response is None:
    print("❌ אין תשובה מהשרת - ננסה שוב מאוחר יותר")
elif response.status_code == 200:
    print(f"קוד סטטוס: {response.status_code}")
    print("✅ הבקשה הצליחה!")
    print(f"מזג האוויר: {response.text}")
else:
    print(f"❌ משהו השתבש. קוד: {response.status_code}")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בואו נראה גם איך נראית שגיאה "אמיתית". השירות httpbin.org מאפשר לבקש ממנו קוד סטטוס לפי בחירה -
נבקש ממנו להעמיד פנים שהדף לא נמצא (404):
</p>

```python
response = safe_get('https://httpbin.org/status/404')

if response is None:
    print("❌ אין תשובה מהשרת")
elif response.status_code == 200:
    print("✅ הבקשה הצליחה!")
else:
    print(f"❌ הדף לא נמצא. קוד: {response.status_code}")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
קיבלנו תשובה מהשרת (לכן <code>response</code> אינו <code>None</code>), אבל הקוד שלה הוא 404 ולא 200.<br>
זו הסיבה שחשוב לבדוק את <code>status_code</code> לפני שמשתמשים במידע.
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">פרמטרים בבקשה (Query Parameters)</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
לפעמים אנחנו רוצים לשלוח פרטים נוספים עם הבקשה.<br>
זה כמו למלא טופס - אנחנו מעבירים מידע שעוזר לשרת לתת לנו בדיוק מה שאנחנו רוצים.<br><br>

דוגמה: נבקש תחזית מפורטת יותר:

```python
# דרך 1: להוסיף את הפרמטרים בכתובת עצמה
response = safe_get('https://wttr.in/Haifa?format=3&lang=he')
print("דרך 1:")
if response is not None:
    print(response.text)
```

## <p style="text-align: right; direction: rtl; float: right; clear: both;">Open-Meteo - שירות מזג אוויר אמין (עם קואורדינטות)</p>

הוא לא מבין שמות של ערים - הוא רוצה <strong>קו רוחב (latitude)</strong> ו<strong>קו אורך (longitude)</strong>.<br>
אל דאגה, הכנו לכם טבלה של חמש ערים בישראל:
</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
נשמור את הטבלה במילון פייתון, ונשלח בקשה עם <code>params</code>.<br>בגרסת האופליין, אם השירות לא זמין הקוד טוען תשובה שמורה מהקובץ <code>resources/weather_sample.json</code> (זה מזג האוויר של תל אביב ביום שהכנו את הקורס) - כך שהשיעור ממשיך גם בלי אינטרנט.<br><strong>זו עקיפה כיתתית בלבד.</strong> בקוד אמיתי לא טוענים קובץ במקום התשובה - פשוט מטפלים בשגיאה: מדפיסים הודעה או מנסים שוב.<br><br>🔍 <strong>שימו לב - זה החלק החשוב:</strong> אנחנו לא יודעים מראש איך התשובה של Open-Meteo בנויה, ואף אחד לא מצפה מכם לדעת. הצעד הראשון עם כל API חדש הוא <strong>להדפיס את התשובה ולקרוא אותה</strong>, ולמצוא בה בעצמכם איפה נמצאת הטמפרטורה. התא הבא מדפיס את התשובה המלאה, ורק אחר כך שולף ממנה את הטמפרטורה.<br>רוצים לראות איך תשובה כזו נראית? פתחו את הקובץ <code>../../resources/weather_sample.json</code> ב-Jupyter או בכל עורך טקסט. זו רק <strong>דוגמה לאיך נראית תשובה</strong>, ולא משהו שטוענים בקוד.
</p>

```python
import json

OPEN_METEO_URL = 'https://api.open-meteo.com/v1/forecast'

# עיר -> (קו רוחב, קו אורך)
CITIES = {
    'תל אביב': (32.0853, 34.7818),
    'ירושלים': (31.7683, 35.2137),
    'חיפה': (32.7940, 34.9896),
    'באר שבע': (31.2530, 34.7915),
    'אילת': (29.5577, 34.9519),
}

lat, lon = CITIES['תל אביב']
params = {
    'latitude': lat,
    'longitude': lon,
    'current': 'temperature_2m,wind_speed_10m',  # מה אנחנו רוצים לדעת עכשיו
    'timezone': 'auto',
}

response = safe_get(OPEN_METEO_URL, params=params)

if response is not None and response.status_code == 200:
    data = response.json()  # הופך את התשובה למילון פייתון (נרחיב על כך במחברת הבאה)
else:
    print("⚠️ משתמשים בנתוני דוגמה שמורים (גרסת אופליין בלבד)")
    with open('../../resources/weather_sample.json', encoding='utf-8') as f:
        data = json.load(f)

if data:
    # צעד ראשון עם כל API חדש: מדפיסים את התשובה וקוראים אותה
    print("התשובה המלאה:")
    print(data)
    print()

    # אחרי שקראנו אותה רואים שהטמפרטורה והרוח נמצאות בתוך המפתח 'current'
    print("הטמפרטורה בתל אביב:", data['current']['temperature_2m'], "מעלות")
    print("מהירות הרוח:", data['current']['wind_speed_10m'], "קמ\"ש")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 <code>CITIES</code> - מילון שהמפתח שלו הוא שם העיר, והערך הוא זוג מספרים (קו רוחב, קו אורך)<br>
🔹 הפרמטרים נשלחו דרך <code>params</code>, ופייתון בנה בשבילנו את הכתובת המלאה<br>
🔹 <code>print(data)</code> - הדפסנו את כל התשובה כדי לראות איך היא בנויה. כך מגלים שהמידע שאנחנו מחפשים נמצא "בתוך" מילון אחר, ולכן כותבים <code>data['current']['temperature_2m']</code><br>
🔹 <strong>רק בגרסת האופליין:</strong> אם הבקשה נכשלה, קראנו את הקובץ השמור במקום. זה נקרא <strong>fallback</strong>, והוא עקיפה כיתתית בלבד - בקוד אמיתי מטפלים בשגיאה (מדפיסים הודעה או מנסים שוב)
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">דוגמה מעשית: בודק מזג אוויר לערים שונות</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
עכשיו נעבור על כל הערים בטבלה בלולאה, ונדפיס את הטמפרטורה של כל אחת.<br>
בין בקשה לבקשה נעשה הפסקה קצרה עם <code>time.sleep(1)</code> - הפקודה אומרת לפייתון "חכה שנייה אחת ואז תמשיך".<br>
למה? כי שרתים לא אוהבים שמציפים אותם בבקשות, ושירות שמרגיש הצפה עלול לחסום אותנו.<br><br>
🛑 <strong>שימו לב:</strong> התא הבא שולח 5 בקשות. הריצו אותו פעם אחת בלבד, ולא שוב ושוב -
כל הכיתה משתמשת באותו שירות, ואם כולנו נריץ בלולאה נקבל חסימה זמנית.
</p>

```python
import time

print("דו״ח מזג אוויר לערים בישראל")
print("=" * 40)

for city_name, (lat, lon) in CITIES.items():
    params = {'latitude': lat, 'longitude': lon, 'current': 'temperature_2m', 'timezone': 'auto'}
    response = safe_get(OPEN_METEO_URL, params=params)

    if response is not None and response.status_code == 200:
        temp = response.json()['current']['temperature_2m']
        print(f"{city_name}: {temp} מעלות")
    else:
        print(f"❌ לא הצלחתי לקבל מידע על {city_name}")

    time.sleep(1)  # הפסקה של שנייה בין בקשות - נימוס בסיסי

print("=" * 40)
print("סיימנו! ✅")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 <code>CITIES.items()</code> נותן לנו בכל סיבוב את שם העיר ואת הזוג (קו רוחב, קו אורך)<br>
🔹 <code>time.sleep(1)</code> גרם לתוכנית לעצור שנייה בכל סיבוב, ולכן ההרצה לקחה כ-5 שניות<br>
🔹 אם עיר אחת נכשלה - ממשיכים לעיר הבאה ולא קורסים
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">Headers - כותרות הבקשה</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
לפעמים השרת רוצה לדעת מי שולח את הבקשה, או איזה סוג מידע אנחנו רוצים לקבל.<br>
Headers זה כמו "מעטפה" שעליה כתוב פרטים על השולח והמכתב.
</p>

```python
# נוסיף Headers שמזהים אותנו
headers = {
    'User-Agent': 'Python Course - Weather App',
    'Accept-Language': 'he-IL'
}

# httpbin.org מחזיר לנו בחזרה את מה שהוא ראה בבקשה שלנו
response = safe_get('https://httpbin.org/headers', headers=headers)

if response is not None and response.status_code == 200:
    seen = response.json()['headers']
    print("השרת ראה את ה-User-Agent שלנו:", seen['User-Agent'])
    print("השרת ראה את השפה שביקשנו:", seen['Accept-Language'])
```
