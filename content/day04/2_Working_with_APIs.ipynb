## <p style="text-align: right; direction: rtl; float: right; clear: both;">מה זה API?</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>API</strong> זה קיצור של Application Programming Interface.<br>
במילים פשוטות: זה "תפריט" שמאפשר לתוכנות לדבר אחת עם השנייה.<br><br>

דוגמה מהחיים:<br>
🔹 אתם הולכים למסעדה (האפליקציה שלכם)<br>
🔹 המלצר מביא תפריט (ה-API)<br>
🔹 אתם מזמינים ממנו (שולחים בקשה)<br>
🔹 המטבח מכין (השרת עובד)<br>
🔹 המלצר מביא לכם את האוכל (התשובה)<br><br>

אתם לא צריכים לדעת איך המטבח עובד - רק מה לבקש איך!
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">מה זה JSON?</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>JSON</strong> (JavaScript Object Notation) זה פורמט לאחסון והעברת מידע.<br>
הוא נראה מאוד דומה למילונים ורשימות בפייתון!<br><br>

דוגמה ל-JSON:
</p>

```python
# זה JSON (כתוב כמחרוזת):
json_example = '''
{
    "name": "דני",
    "age": 30,
    "city": "תל אביב",
    "hobbies": ["ספורט", "קריאה", "בישול"]
}
'''

print("JSON לדוגמה:")
print(json_example)
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
שימו לב שזה דומה למילון פייתון, אבל יש הבדלים קטנים:<br>
🔹 ב-JSON משתמשים ב-<code>"</code> (גרשיים כפולים) תמיד<br>
🔹 אין <code>True/False</code> אלא <code>true/false</code> (אותיות קטנות)<br>
🔹 JSON זה מחרוזת, צריך להמיר אותו למילון פייתון<br>
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">המרת JSON למילון פייתון</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
המודול <code>json</code> של פייתון יודע להפוך מחרוזת JSON למילון אמיתי עם <code>json.loads()</code>
(ה-s בסוף הוא מ-string).<br>
אחרי ההמרה אפשר לעבוד איתו בדיוק כמו עם כל מילון.
</p>

```python
import json

json_string = '{"name": "שרה", "age": 25, "city": "חיפה"}'

# המרה ל-dictionary של פייתון
person = json.loads(json_string)

print("סוג המשתנה:", type(person))
print("התוכן:", person)
print()
print("גישה לערכים:")
print(f"שם: {person['name']}")
print(f"גיל: {person['age']}")
print(f"עיר: {person['city']}")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 <code>json.loads()</code> לקח מחרוזת והחזיר מילון פייתון<br>
🔹 את הערכים שולפים בעזרת המפתח, בדיוק כמו תמיד: <code>person['name']</code>
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">קריאה ל-API עם תשובת JSON</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בואו נשתמש ב-API של ציטוטים מפורסמים (dummyjson.com):<br>
זה API חינמי, בלי מפתח, שנותן לנו ציטוט אקראי.<br>
כל ציטוט מגיע עם שלושה שדות: <code>id</code> (מספר), <code>quote</code> (הטקסט) ו-<code>author</code> (שם המחבר).<br>
שימו לב: הציטוטים באנגלית.<br><br>
כדי שהשיעור לא ייתקע אם השירות לא זמין, הכנו בקובץ <code>resources/quotes_sample.json</code> כמה ציטוטים שמורים לגיבוי. נטען אותם קודם:
</p>

```python
import requests
import json

# ציטוטים שמורים - לגיבוי למקרה שאין אינטרנט
with open('../../resources/quotes_sample.json', encoding='utf-8') as f:
    SAMPLE_QUOTES = json.load(f)

QUOTE_URL = 'https://dummyjson.com/quotes/random'

print(f"נטענו {len(SAMPLE_QUOTES)} ציטוטים שמורים")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
עכשיו נבקש ציטוט אקראי מהשרת. שימו לב ל-<code>timeout=10</code> ול-<code>try / except</code>:
</p>

```python
# נבקש ציטוט אקראי
try:
    response = requests.get(QUOTE_URL, timeout=10)

    print("סטטוס:", response.status_code)
    print()
    print("התשובה כטקסט:")
    print(response.text)
    print()
    print("סוג התשובה:", type(response.text))
except requests.exceptions.RequestException:
    print("⚠️ לא הצלחנו להתחבר לשירות. ממשיכים עם הציטוטים השמורים.")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
אוקיי, קיבלנו JSON אבל זה עדיין טקסט (<code>str</code>). בואו נהפוך אותו למילון:
</p>

```python
# הדרך הארוכה (עם json.loads)
try:
    response = requests.get(QUOTE_URL, timeout=10)
    data = json.loads(response.text)
except Exception:
    print("⚠️ משתמשים בציטוט שמור")
    data = SAMPLE_QUOTES[0]

print("הציטוט:")
print(f'"{data["quote"]}"')
print(f"- {data['author']}")
```

```python
# הדרך הקצרה והנוחה (מומלץ!)
try:
    response = requests.get(QUOTE_URL, timeout=10)
    data = response.json()  # 👈 requests עושה את ההמרה בשבילנו!
except Exception:
    print("⚠️ משתמשים בציטוט שמור")
    data = SAMPLE_QUOTES[1]

print("ציטוט אקראי:")
print(f'"{data["quote"]}"')
print(f"- {data['author']}")
print()
print(f"אורך: {len(data['quote'])} תווים")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 <code>response.text</code> הוא מחרוזת. <code>response.json()</code> הוא מילון מוכן לשימוש<br>
🔹 את האורך חישבנו בעצמנו עם <code>len()</code> - ה-API לא מחזיר אותו<br>
🔹 בכל פעם שמריצים את התא מקבלים ציטוט אחר, כי השרת בוחר אקראית<br>
🔹 אם יש תקלה, ה-<code>except</code> טוען ציטוט שמור במקום, והשיעור ממשיך
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">טיפול בשגיאות - חשוב מאוד!</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
לא תמיד הכל עובד חלק. אולי אין אינטרנט, אולי השרת נפל, אולי הכתובת שגויה.<br>
חשוב לטפל בשגיאות כדי שהתוכנה שלנו לא תקרוס! 🛡️
</p>

```python
# דוגמה 1: בדיקת סטטוס קוד
url = 'https://dog.ceo/api/breed/unicorn/images/random'  # גזע שלא קיים

try:
    response = requests.get(url, timeout=10)

    if response.status_code == 200:
        data = response.json()
        print("הכל עבד!")
        print(data['message'])
    else:
        print(f"❌ שגיאה! קוד סטטוס: {response.status_code}")
        print("התשובה מהשרת:", response.text)
except requests.exceptions.RequestException:
    print("❌ בעיית רשת - אי אפשר להגיע לשרת")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
אין גזע כלבים שנקרא "חד קרן", ולכן השרת ענה 404.<br>
זו שגיאה "נורמלית" - השרת ענה לנו, אבל בלי המידע שביקשנו. לכן בודקים את <code>status_code</code>.
</p>

```python
# דוגמה 2: טיפול בשגיאות רשת (אין אינטרנט, השרת לא עונה)
try:
    response = requests.get(QUOTE_URL, timeout=10)  # חכה מקסימום 10 שניות

    if response.status_code == 200:
        data = response.json()
        print(f'"{data["quote"]}"')
        print(f"- {data['author']}")
    else:
        print(f"השרת החזיר שגיאה: {response.status_code}")

except requests.exceptions.Timeout:
    print("❌ הבקשה לקחה יותר מדי זמן (timeout)")
except requests.exceptions.ConnectionError:
    print("❌ לא הצלחתי להתחבר לשרת - בדקו את חיבור האינטרנט")
except Exception as e:
    print(f"❌ קרתה שגיאה לא צפויה: {e}")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
יש כמה סוגים של תקלות, ואפשר להגיב לכל סוג בנפרד:<br>
🔹 <code>Timeout</code> - השרת לא ענה בזמן<br>
🔹 <code>ConnectionError</code> - אין חיבור לאינטרנט, או שהשרת לא קיים<br>
🔹 <code>Exception</code> - כל תקלה אחרת שלא חשבנו עליה
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">פונקציה נוחה לבקשות API</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בואו נכתוב פונקציה שמטפלת בכל השגיאות בשבילנו, פעם אחת, ואז נשתמש בה בכל מקום.<br>
היא מחזירה את המילון מהתשובה, או <code>None</code> אם משהו השתבש:
</p>

```python
def get_api_data(url, params=None, timeout=10):
    """
    פונקציה לשליחת בקשת GET ל-API עם טיפול בשגיאות

    Parameters:
    url (str): כתובת ה-API
    params (dict): פרמטרים לבקשה (אופציונלי)
    timeout (int): זמן המתנה מקסימלי בשניות

    Returns:
    dict: המידע מה-API או None במקרה של שגיאה
    """
    try:
        response = requests.get(url, params=params, timeout=timeout)

        if response.status_code == 200:
            return response.json()
        else:
            print(f"❌ קוד שגיאה: {response.status_code}")
            return None

    except requests.exceptions.Timeout:
        print("❌ הבקשה לקחה יותר מדי זמן")
        return None
    except requests.exceptions.ConnectionError:
        print("❌ בעיה בחיבור לאינטרנט")
        return None
    except Exception as e:
        print(f"❌ שגיאה: {e}")
        return None


# נסיון של הפונקציה
print("דוגמה: ציטוט")
data = get_api_data(QUOTE_URL)
if data:
    print(f'"{data["quote"]}" - {data["author"]}')
else:
    print("(אין תשובה כרגע - ממשיכים הלאה)")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 הפונקציה עוטפת את כל הטיפול בשגיאות במקום אחד<br>
🔹 מי שמשתמש בה צריך לבדוק רק דבר אחד: <code>if data:</code> (האם קיבלנו מידע?)<br>
🔹 כך אנחנו לא חוזרים על אותו <code>try / except</code> בכל מקום
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">API עם פרמטרים - תמונות של כלבים! 🐕</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
ה-API של dog.ceo נותן לנו תמונות אקראיות של כלבים.<br>
אפשר גם לבקש גזעים ספציפיים!
</p>

```python
# תמונה אקראית של כלב
data = get_api_data('https://dog.ceo/api/breeds/image/random')

if data:
    print("סטטוס הבקשה:", data['status'])
    print("לינק לתמונה:")
    print(data['message'])
    print()
    print("העתיקו את הלינק לדפדפן כדי לראות את הכלב! 🐕")
```

```python
# קבלת מספר תמונות של כלבים
data = get_api_data('https://dog.ceo/api/breeds/image/random/3')

if data:
    print("3 תמונות אקראיות של כלבים:")
    print("=" * 50)
    for i, image_url in enumerate(data['message'], 1):
        print(f"{i}. {image_url}")
```

```python
# בואו נראה אילו גזעים יש
data = get_api_data('https://dog.ceo/api/breeds/list/all')

if data:
    breeds = list(data['message'].keys())
    print(f"יש {len(breeds)} גזעים!")
    print()
    print("דוגמאות לגזעים:")
    for breed in breeds[:10]:  # רק 10 הראשונים
        print(f"- {breed}")
```

```python
# תמונה של גזע ספציפי
breed = 'husky'
data = get_api_data(f'https://dog.ceo/api/breed/{breed}/images/random')

if data and data['status'] == 'success':
    print(f"תמונה של {breed}:")
    print(data['message'])
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 לפעמים התשובה היא מילון עם שדה <code>message</code> שמכיל מחרוזת, ולפעמים רשימה, או מילון של גזעים<br>
🔹 כדי לדעת מה יש בפנים, מדפיסים את התשובה ומסתכלים על המבנה שלה<br>
🔹 כתובת עם <code>/random/3</code> מחזירה 3 תמונות במקום אחת - הפרמטר הוא חלק מהכתובת
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">הורדת תמונה</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
עד עכשיו קיבלנו מה-API רק <strong>כתובת</strong> של תמונה. בואו נוריד את התמונה עצמה ונשמור אותה על המחשב!<br><br>
תמונה היא לא טקסט - היא רצף של בתים (bytes). לכן:<br>
🔹 במקום <code>response.text</code> משתמשים ב-<code>response.content</code> (הבתים הגולמיים)<br>
🔹 פותחים את הקובץ במצב <code>'wb'</code> - <strong>w</strong>rite <strong>b</strong>inary, כתיבה בינארית<br>
🔹 את התמונה שומרים בתיקייה, ואם היא לא קיימת יוצרים אותה עם <code>os.makedirs(..., exist_ok=True)</code><br><br>
קודם נוריד את הבתים ונבדוק מה קיבלנו:
</p>

```python
image_bytes = None

data = get_api_data('https://dog.ceo/api/breeds/image/random')

if data:
    image_url = data['message']
    print("כתובת התמונה:", image_url)

    try:
        image_response = requests.get(image_url, timeout=10)
        image_bytes = image_response.content  # התוכן הבינארי של התמונה
        print("סוג המידע:", type(image_bytes))
        print("גודל התמונה בבתים:", len(image_bytes))
    except requests.exceptions.RequestException:
        print("❌ ההורדה נכשלה")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
עכשיו נשמור את הבתים לקובץ. <code>os.makedirs</code> יוצר את התיקייה <code>downloads</code>
(וה-<code>exist_ok=True</code> אומר "אל תיכשל אם היא כבר קיימת" - כך אפשר להריץ שוב ושוב):
</p>

```python
import os

os.makedirs('downloads', exist_ok=True)

if image_bytes:
    with open('downloads/dog_example.jpg', 'wb') as f:
        f.write(image_bytes)

    print("התמונה נשמרה! ✅")
    print("הקבצים בתיקייה downloads:", os.listdir('downloads'))
else:
    print("אין תמונה לשמור - הריצו שוב את התא הקודם")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
ואפשר גם להציג את התמונה ישר במחברת:
</p>

```python
from IPython.display import Image, display

if image_bytes:
    display(Image(data=image_bytes, width=300))
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 <code>response.content</code> החזיר בתים (<code>bytes</code>) ולא טקסט<br>
🔹 <code>open(..., 'wb')</code> פתח קובץ לכתיבה בינארית. עם <code>'w'</code> רגיל פייתון היה מנסה לכתוב טקסט ולא היה מצליח לכתוב תמונה<br>
🔹 <code>os.makedirs('downloads', exist_ok=True)</code> דאג שהתיקייה קיימת<br>
🔹 בהורדה של הרבה תמונות, פשוט חוזרים על אותם שלבים בתוך לולאה ונותנים לכל קובץ שם אחר (<code>dog_1.jpg</code>, <code>dog_2.jpg</code>...)
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">JSON מקונן - מילון בתוך מילון</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בעולם האמיתי, תשובות של APIs הן לא "שטוחות". לרוב יש מילון שבתוכו מילון נוסף, או מילון שבתוכו רשימה.<br>
זה נקרא JSON <strong>מקונן</strong> (nested), והשיטה פשוטה: יורדים שלב אחרי שלב, עם סוגריים מרובעים אחד אחרי השני.<br><br>
נתחיל בדוגמה: פרטי משתמש מהשירות jsonplaceholder (נתוני בדיקה, לא אנשים אמיתיים).
שימו לב שהכתובת והמיקום הם מילונים בתוך המילון:
</p>

```python
user = get_api_data('https://jsonplaceholder.typicode.com/users/1')

if user:
    # נדפיס יפה את כל המבנה, כדי לראות מה יש בפנים
    print(json.dumps(user, indent=2, ensure_ascii=False))
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
עכשיו נשלוף ערכים משלבים שונים:
</p>

```python
if user:
    print("שם:", user['name'])                       # שלב 1
    print("עיר:", user['address']['city'])           # שלב 2: address ואז city
    print("קו רוחב:", user['address']['geo']['lat'])  # שלב 3: address ואז geo ואז lat
    print("שם החברה:", user['company']['name'])
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 <code>json.dumps(..., indent=2)</code> מדפיס את המילון בצורה מסודרת עם הזחות - טריק מצוין כדי "להבין" תשובה חדשה<br>
🔹 <code>ensure_ascii=False</code> גורם לעברית להופיע כמו שצריך<br>
🔹 <code>user['address']['city']</code> - קודם מוציאים את המילון <code>address</code>, ומתוכו את <code>city</code><br><br>
המקרה השני הוא <strong>רשימה בתוך מילון</strong>. כך נראית תחזית הימים הקרובים ב-Open-Meteo:
המפתח <code>daily</code> מכיל מילון, ובו כל מפתח מכיל רשימה (ערך לכל יום).
</p>

```python
OPEN_METEO_URL = 'https://api.open-meteo.com/v1/forecast'
params = {
    'latitude': 32.0853,   # תל אביב
    'longitude': 34.7818,
    'daily': 'temperature_2m_max,temperature_2m_min',
    'timezone': 'auto',
    'forecast_days': 4,
}

weather = get_api_data(OPEN_METEO_URL, params=params)

if weather is None:
    print("⚠️ משתמשים בנתוני דוגמה שמורים")
    with open('../../resources/weather_sample.json', encoding='utf-8') as f:
        weather = json.load(f)

days = weather['daily']['time']                # רשימת תאריכים
highs = weather['daily']['temperature_2m_max']  # רשימת טמפרטורות מקסימום
lows = weather['daily']['temperature_2m_min']   # רשימת טמפרטורות מינימום

print("תחזית ל-4 ימים בתל אביב:")
for i in range(len(days)):
    print(f"{days[i]}: בין {lows[i]} ל-{highs[i]} מעלות")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 <code>weather['daily']</code> הוא מילון, ו-<code>weather['daily']['time']</code> היא רשימה<br>
🔹 הרשימות מסודרות באותו סדר: מקום 0 הוא היום, מקום 1 הוא מחר, וכן הלאה<br>
🔹 לכן בלולאה פונים לאותו אינדקס <code>i</code> בכל שלוש הרשימות<br><br>
והמקרה השלישי: <strong>רשימה של מילונים</strong>. השירות dummyjson מחזיר מוצרים, וכל מוצר הוא מילון:
</p>

```python
data = get_api_data('https://dummyjson.com/products', params={'limit': 3, 'select': 'title,price'})

if data:
    print("סה\"כ מוצרים בשרת:", data['total'])
    for product in data['products']:   # data['products'] היא רשימה של מילונים
        print(f"- {product['title']}: ${product['price']}")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 <code>data['products']</code> היא רשימה, ולכן עוברים עליה בלולאה<br>
🔹 בכל סיבוב <code>product</code> הוא מילון, ושולפים ממנו ערכים לפי מפתח<br><br>
<strong>כלל אצבע:</strong> רואים <code>{ }</code> - פונים עם שם מפתח. רואים <code>[ ]</code> - פונים עם מספר או עוברים בלולאה.
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">דוגמה מורכבת: יוצרים אוסף ציטוטים</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
נשתמש בפונקציה שכתבנו כדי לאסוף כמה ציטוטים ולשמור אותם ברשימה של מילונים.<br>
אם אחת הבקשות נכשלה, נשתמש בציטוט שמור כדי שהרשימה תהיה מלאה:
</p>

```python
def collect_quotes(num_quotes=5):
    """
    אוספת מספר ציטוטים מ-API
    """
    quotes = []

    print(f"אוסף {num_quotes} ציטוטים...")
    print("=" * 60)

    for i in range(num_quotes):
        data = get_api_data(QUOTE_URL)
        if data is None:
            print("⚠️ משתמשים בציטוט שמור")
            data = SAMPLE_QUOTES[i % len(SAMPLE_QUOTES)]

        quote_info = {
            'text': data['quote'],
            'author': data['author'],
            'length': len(data['quote'])
        }
        quotes.append(quote_info)
        print(f"\n{i+1}. \"{quote_info['text']}\"")
        print(f"   - {quote_info['author']} ({quote_info['length']} תווים)")

    return quotes


# נאסוף 3 ציטוטים
my_quotes = collect_quotes(3)
print("\n" + "=" * 60)
print(f"סה\"כ נאספו {len(my_quotes)} ציטוטים")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 הפונקציה החזירה <strong>רשימה של מילונים</strong>, שאפשר לשמור לקובץ JSON או CSV (כמו שלמדנו ביום 3)<br>
🔹 <code>i % len(SAMPLE_QUOTES)</code> מבטיח שלא נחרוג מגבולות הרשימה השמורה
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">תרגיל: חקר API</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בחרו אחד מה-APIs למעלה וכתבו פונקציה שמקבלת פרמטר ומחזירה מידע מעניין.<br>
למשל: פונקציה שמקבלת שם גזע ומחזירה תמונה, או פונקציה שמקבלת מספר מוצר ומחזירה את שמו ומחירו (<code>https://dummyjson.com/products/1</code>).
</p>

```python
# דוגמה: פונקציה שמחזירה תמונה של גזע כלב לפי בחירה
def get_dog_by_breed(breed):
    """
    מחזירה תמונה של כלב מגזע מסוים
    """
    url = f'https://dog.ceo/api/breed/{breed}/images/random'
    data = get_api_data(url)

    if data and data['status'] == 'success':
        return data['message']
    else:
        return None

# נסיון
image_url = get_dog_by_breed('corgi')
if image_url:
    print(f"תמונה של קורגי: {image_url}")
else:
    print("לא מצאתי את הגזע הזה")

# עכשיו התור שלכם! נסו ליצור פונקציה משלכם
```
