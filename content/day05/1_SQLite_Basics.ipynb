## <p style="text-align: right; direction: rtl; float: right; clear: both;">מה זה SQL?</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
SQL (Structured Query Language) היא שפה מיוחדת לעבודה עם מסדי נתונים.<br>
היא מאפשרת לנו:<br>
✅ ליצור טבלאות<br>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">התחברות למסד נתונים</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בואו נתחבר למסד נתונים. אם הקובץ לא קיים, SQLite ייצור אותו אוטומטית.<br>
כדי שאפשר יהיה להריץ את המחברת שוב ושוב, התא הבא מתחיל בכך שהוא סוגר חיבור ישן (אם יש) ומוחק את קובץ המסד הישן.
כך כל הרצה מתחילה מאפס.
</p>

```python
import sqlite3
import os

# אם כבר הרצנו את התא הזה בעבר - סוגרים את החיבור הישן ומוחקים את הקובץ הישן
try:
    connection.close()
except NameError:
    pass  # זו ההרצה הראשונה, עוד אין חיבור ישן

if os.path.exists('my_database.db'):
    os.remove('my_database.db')

# יצירת חיבור למסד נתונים
# אם הקובץ לא קיים, הוא ייווצר אוטומטית
connection = sqlite3.connect('my_database.db')

# יצירת cursor - כלי לביצוע פקודות SQL
cursor = connection.cursor()

print("התחברנו בהצלחה למסד הנתונים! 🎉")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 <code>sqlite3.connect('my_database.db')</code> יצר קובץ מסד נתונים חדש בתיקייה של המחברת ופתח אליו "דלת" (חיבור)<br>
🔹 <code>connection.cursor()</code> יצר "עט" שבעזרתו נכתוב פקודות SQL<br>
🔹 <code>os.remove</code> מחק את הקובץ הישן, כדי שהרצה חוזרת לא תוסיף שורות כפולות
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">יצירת טבלה - CREATE TABLE</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בואו ניצור טבלה לשמירת פרטי תלמידים.<br>
כל תלמיד יהיה עם: מזהה, שם, גיל ועיר.
</p>

```python
# יצירת טבלת תלמידים
cursor.execute('''
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    age INTEGER,
    city TEXT
)
''')

# שמירת השינויים
connection.commit()

print("הטבלה נוצרה בהצלחה! ✅")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
הסבר על הפקודה:<br>
📌 <strong>CREATE TABLE</strong> - יוצר טבלה חדשה<br>
📌 <strong>IF NOT EXISTS</strong> - יוצר רק אם הטבלה לא קיימת (למנוע שגיאות)<br>
📌 <strong>INTEGER PRIMARY KEY AUTOINCREMENT</strong> - מזהה ייחודי שעולה אוטומטית<br>
📌 <strong>TEXT</strong> - טיפוס טקסט (כמו str בפייתון)<br>
📌 <strong>NOT NULL</strong> - שדה חובה (לא יכול להיות ריק)
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">הוספת נתונים - INSERT</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
עכשיו בואו נוסיף תלמידים לטבלה. נתחיל בתלמיד אחד:
</p>

```python
# הוספת תלמיד אחד
cursor.execute('''
INSERT INTO students (name, age, city)
VALUES ('דני כהן', 25, 'תל אביב')
''')

connection.commit()
print("תלמיד אחד נוסף! ✅")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
ועכשיו כמה תלמידים בבת אחת עם <code>executemany</code>. שימו לב לסימני השאלה <code>?</code> -
מחליפים אותם בערכים מתוך הרשימה, שורה אחרי שורה:
</p>

```python
# הוספת מספר תלמידים בבת אחת
students_data = [
    ('שרה לוי', 23, 'חיפה'),
    ('יוסי כהן', 27, 'ירושלים'),
    ('רחל אברהם', 22, 'באר שבע'),
    ('משה דוד', 26, 'תל אביב')
]

cursor.executemany('''
INSERT INTO students (name, age, city)
VALUES (?, ?, ?)
''', students_data)

connection.commit()
print(f"נוספו {len(students_data)} תלמידים! 🎉")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
💡 <strong>טיפ:</strong> הסימן <code>?</code> משמש כ-placeholder (מציין מיקום) לערכים.<br>
זה בטוח יותר ומונע התקפות SQL Injection.
</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 <code>execute</code> הוסיפה שורה אחת, <code>executemany</code> הוסיפה ארבע שורות בפקודה אחת<br>
🔹 <code>connection.commit()</code> שומר את השינויים בקובץ. בלי commit השינויים נעלמים כשסוגרים את החיבור!<br>
🔹 אם מריצים את אחד מתאי ה-INSERT פעם נוספת בלי להתחיל מהתחלה, השורות יתווספו שוב. כדי לנקות - הריצו מחדש מתא "התחברות למסד נתונים"
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">שליפת נתונים - SELECT</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
עכשיו נלמד לשלוף ולהציג את המידע שהכנסנו:
</p>

```python
# שליפת כל התלמידים
cursor.execute('SELECT * FROM students')

# קבלת כל השורות
all_students = cursor.fetchall()

print("כל התלמידים:")
for student in all_students:
    print(f"ID: {student[0]}, שם: {student[1]}, גיל: {student[2]}, עיר: {student[3]}")
```

```python
# שליפת עמודות ספציפיות בלבד
cursor.execute('SELECT name, city FROM students')

results = cursor.fetchall()
print("\nשמות וערים:")
for name, city in results:
    print(f"{name} מ{city}")
```

```python
# שליפת תלמיד אחד בלבד
cursor.execute('SELECT * FROM students LIMIT 1')

first_student = cursor.fetchone()
print(f"\nהתלמיד הראשון: {first_student}")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 <code>SELECT *</code> מביא את כל העמודות; <code>SELECT name, city</code> מביא רק את מה שביקשנו<br>
🔹 <code>fetchall()</code> מחזיר רשימה של כל השורות; כל שורה היא tuple (למשל <code>(1, 'דני כהן', 25, 'תל אביב')</code>)<br>
🔹 <code>fetchone()</code> מחזיר שורה אחת בלבד<br>
🔹 <code>LIMIT 1</code> אומר ל-SQL "תחזיר לכל היותר שורה אחת"
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">סינון נתונים - WHERE</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
WHERE מאפשר לנו לסנן את התוצאות לפי תנאים:<br>
</p>

```python
# תלמידים מתל אביב בלבד
cursor.execute('''
SELECT name, age FROM students
WHERE city = 'תל אביב'
''')

tel_aviv_students = cursor.fetchall()
print("תלמידים מתל אביב:")
for name, age in tel_aviv_students:
    print(f"{name}, גיל {age}")
```

```python
# תלמידים מעל גיל 25
cursor.execute('''
SELECT name, age FROM students
WHERE age > 25
''')

older_students = cursor.fetchall()
print("\nתלמידים מעל גיל 25:")
for name, age in older_students:
    print(f"{name}, גיל {age}")
```

```python
# שילוב של מספר תנאים
cursor.execute('''
SELECT name, age, city FROM students
WHERE age >= 25 AND city = 'תל אביב'
''')

filtered_students = cursor.fetchall()
print("\nתלמידים מתל אביב מעל גיל 25:")
for name, age, city in filtered_students:
    print(f"{name}, גיל {age}, {city}")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 <code>WHERE</code> משאיר רק שורות שעומדות בתנאי<br>
🔹 אפשר לחבר תנאים עם <code>AND</code> (שניהם חייבים להתקיים) או <code>OR</code> (מספיק אחד)<br>
🔹 שימו לב: טקסט ב-SQL כותבים בגרשיים <strong>בודדים</strong>: <code>'תל אביב'</code>
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">אופרטורים ב-WHERE</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
לצד <code>=</code>, <code>&gt;</code> ו-<code>&lt;</code> יש שלושה אופרטורים שימושיים:<br>
🔹 <code>LIKE</code> - חיפוש חלקי בטקסט (הסימן <code>%</code> מייצג "כל דבר")<br>
🔹 <code>IN</code> - האם הערך נמצא ברשימה<br>
🔹 <code>BETWEEN</code> - האם המספר בטווח (כולל הקצוות)
</p>

```python
# LIKE - חיפוש חלקי (כמו contains)
cursor.execute('''
SELECT name FROM students
WHERE name LIKE '%כהן%'
''')

cohens = cursor.fetchall()
print("תלמידים עם שם משפחה 'כהן':")
for (name,) in cohens:
    print(f"- {name}")
```

```python
# IN - בדיקה האם הערך נמצא ברשימה
cursor.execute('''
SELECT name, city FROM students
WHERE city IN ('תל אביב', 'חיפה')
''')

big_cities = cursor.fetchall()
print("\nתלמידים מערים גדולות:")
for name, city in big_cities:
    print(f"{name} - {city}")
```

```python
# BETWEEN - בדיקת טווח
cursor.execute('''
SELECT name, age FROM students
WHERE age BETWEEN 23 AND 26
''')

age_range = cursor.fetchall()
print("\nתלמידים בגילאים 23-26:")
for name, age in age_range:
    print(f"{name}, גיל {age}")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 <code>'%כהן%'</code> = "כל שם שמכיל את המילה כהן בכל מקום"<br>
🔹 <code>IN ('תל אביב', 'חיפה')</code> קצר יותר מ-<code>city = 'תל אביב' OR city = 'חיפה'</code><br>
🔹 <code>BETWEEN 23 AND 26</code> כולל את 23 ואת 26
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">שאילתות עם פרמטרים - WHERE ?</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
עד עכשיו כתבנו את הערכים ישירות בתוך פקודת ה-SQL. אבל בתוכנית אמיתית הערך מגיע ממשתנה - למשל ממה שמשתמש הקליד.<br>
הדרך הנכונה: לכתוב <code>?</code> בפקודה, ולהעביר את הערכים בנפרד בתוך tuple.<br>
שימו לב: tuple עם ערך אחד נכתב עם פסיק בסוף: <code>(city,)</code>.
</p>

```python
# הערך מגיע ממשתנה
city_to_find = 'תל אביב'  # שנו כאן לעיר אחרת

cursor.execute('SELECT name, age FROM students WHERE city = ?', (city_to_find,))
print(f"תלמידים מ{city_to_find}:")
for name, age in cursor.fetchall():
    print(f"- {name}, גיל {age}")

# אפשר כמה פרמטרים, לפי הסדר
min_age, max_age = 23, 26
cursor.execute('SELECT name, age FROM students WHERE age BETWEEN ? AND ?', (min_age, max_age))
print(f"\nתלמידים בגילאים {min_age}-{max_age}:")
for name, age in cursor.fetchall():
    print(f"- {name}, גיל {age}")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 הסימן <code>?</code> שומר מקום; הערכים נכנסים אליו לפי הסדר מתוך ה-tuple<br>
🔹 אפשר לשנות את המשתנה ולהריץ שוב בלי לגעת בפקודת ה-SQL<br><br>
<strong>למה לא פשוט f-string?</strong> כי אז מישהו זדוני יכול "להכניס" פקודת SQL בתוך הטקסט. זה נקרא SQL Injection, וזו אחת התקלות האבטחה הנפוצות ביותר בעולם. הנה הדגמה (בטוחה - רק קוראת מידע):
</p>

```python
user_input = "x' OR '1'='1"   # טקסט "ערמומי" שמישהו הקליד

# דרך גרועה: בניית הפקודה עם f-string
bad_query = f"SELECT name FROM students WHERE city = '{user_input}'"
print("הפקודה שנבנתה:", bad_query)
cursor.execute(bad_query)
print("תוצאה (דליפה! קיבלנו את כולם):", cursor.fetchall())

# דרך נכונה: פרמטר עם ?
cursor.execute('SELECT name FROM students WHERE city = ?', (user_input,))
print("תוצאה עם ? (אין עיר כזו, אז אין תוצאות):", cursor.fetchall())
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
ב-f-string הטקסט הערמומי הפך את התנאי ל-<code>city = 'x' OR '1'='1'</code> - תנאי שתמיד נכון - ולכן קיבלנו את כל השורות.<br>
עם <code>?</code> פייתון מתייחס לכל מה שהמשתמש הקליד כאל <strong>טקסט בלבד</strong>, ולא כאל פקודה.<br>
כלל: <strong>תמיד משתמשים ב-<code>?</code> כשערך מגיע ממשתנה.</strong>
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">מיון תוצאות - ORDER BY</p>

```python
# מיון לפי גיל (מהצעיר למבוגר)
cursor.execute('''
SELECT name, age FROM students
ORDER BY age
''')

sorted_by_age = cursor.fetchall()
print("תלמידים ממוינים לפי גיל:")
for name, age in sorted_by_age:
    print(f"{name}, גיל {age}")
```

```python
# מיון הפוך (מהמבוגר לצעיר)
cursor.execute('''
SELECT name, age FROM students
ORDER BY age DESC
''')

sorted_desc = cursor.fetchall()
print("\nתלמידים לפי גיל (מהמבוגר לצעיר):")
for name, age in sorted_desc:
    print(f"{name}, גיל {age}")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
ואפשר לשלב מיון עם <code>LIMIT</code> כדי לקבל "טופ" כלשהו - למשל שלושת התלמידים המבוגרים ביותר:
</p>

```python
cursor.execute('SELECT name, age FROM students ORDER BY age DESC LIMIT 3')

print("שלושת המבוגרים ביותר:")
for name, age in cursor.fetchall():
    print(f"{name}, גיל {age}")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 <code>ORDER BY age</code> ממיין מהקטן לגדול; <code>DESC</code> הופך את הסדר<br>
🔹 <code>LIMIT 3</code> חותך אחרי שלוש שורות. קודם ממיינים ואז חותכים, ולכן קיבלנו את שלושת המבוגרים
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">פונקציות צבירה</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
פונקציות צבירה "מסכמות" הרבה שורות לערך אחד: <code>COUNT</code> סופר, <code>AVG</code> מחשב ממוצע,
<code>MIN</code> ו-<code>MAX</code> מוצאים את הקטן והגדול ביותר.
</p>

```python
# COUNT - ספירת שורות
cursor.execute('SELECT COUNT(*) FROM students')
total_students = cursor.fetchone()[0]
print(f"סה\"כ תלמידים: {total_students}")

# AVG - ממוצע
cursor.execute('SELECT AVG(age) FROM students')
avg_age = cursor.fetchone()[0]
print(f"גיל ממוצע: {avg_age:.1f}")

# MIN - מינימום
cursor.execute('SELECT MIN(age) FROM students')
min_age = cursor.fetchone()[0]
print(f"הגיל הצעיר ביותר: {min_age}")

# MAX - מקסימום
cursor.execute('SELECT MAX(age) FROM students')
max_age = cursor.fetchone()[0]
print(f"הגיל המבוגר ביותר: {max_age}")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
כל שאילתה החזירה שורה אחת עם ערך אחד, ולכן השתמשנו ב-<code>fetchone()[0]</code> כדי להוציא אותו.<br>
(בהמשך נראה איך לקבל סיכום לכל קבוצה בנפרד עם GROUP BY.)
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">דוגמה מלאה - מערכת ניהול ספרייה</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בואו נחבר את כל מה שלמדנו לדוגמה אחת: ספרייה קטנה עם ספרים אמיתיים.<br>
זו מערכת נפרדת עם קובץ מסד נתונים משלה (<code>library.db</code>), והתא מתחיל במחיקת הקובץ הישן כדי שאפשר יהיה להריץ אותו שוב.
</p>

```python
# יצירת מסד נתונים לספרייה
try:
    conn.close()
except NameError:
    pass

if os.path.exists('library.db'):
    os.remove('library.db')

conn = sqlite3.connect('library.db')
cursor = conn.cursor()

# יצירת טבלת ספרים
cursor.execute('''
CREATE TABLE books (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    title TEXT NOT NULL,
    author TEXT NOT NULL,
    year INTEGER,
    available INTEGER DEFAULT 1
)
''')

# הוספת ספרים (available: 1 = זמין, 0 = מושאל)
books = [
    ('סיפור על אהבה וחושך', 'עמוס עוז', 2002, 1),
    ('תמול שלשום', 'ש"י עגנון', 1945, 1),
    ('אישה בורחת מבשורה', 'דוד גרוסמן', 2008, 0),
    ('עיין ערך אהבה', 'דוד גרוסמן', 1986, 1),
    ('דירה להשכיר', 'ליאה גולדברג', 1959, 1),
    ('מיכאל שלי', 'עמוס עוז', 1968, 0),
]

cursor.executemany('INSERT INTO books (title, author, year, available) VALUES (?, ?, ?, ?)', books)
conn.commit()

# חיפוש ספרים זמינים
print("ספרים זמינים להשאלה:")
cursor.execute('SELECT title, author FROM books WHERE available = 1')
for title, author in cursor.fetchall():
    print(f"📖 {title} - {author}")

# חיפוש ספרים לפי מחבר
print("\nספרים של דוד גרוסמן:")
cursor.execute("SELECT title, year FROM books WHERE author LIKE '%גרוסמן%'")
for title, year in cursor.fetchall():
    print(f"📚 {title} ({year})")

conn.close()
print("\n✅ הספרייה נוצרה בהצלחה!")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
🔹 <code>available INTEGER DEFAULT 1</code> - ב-SQLite אין טיפוס "כן/לא", אז משתמשים ב-1 ו-0<br>
🔹 את הפקודה עם <code>LIKE</code> כתבנו בגרשיים כפולים של פייתון, ובתוכה גרשיים <strong>בודדים</strong> של SQL - זה הסדר הנכון<br>
🔹 את החיבור <code>conn</code> סגרנו בסוף התא, כי הוא היה שייך רק לתא הזה
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">סגירת החיבור</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
חשוב לסגור את החיבור למסד הנתונים כשסיימנו.<br>
זה התא <strong>האחרון</strong> במחברת - אחריו אי אפשר להריץ שאילתות על <code>connection</code> (תקבלו <code>Cannot operate on a closed database</code>).
אם רוצים לחזור ולהתנסות, מריצים שוב מהתא "התחברות למסד נתונים".
</p>

```python
# סגירת החיבור
connection.close()
print("החיבור למסד הנתונים נסגר בהצלחה! 👋")
```
