## <p style="text-align: right; direction: rtl; float: right; clear: both;">הקדמה: גרף אחד עם matplotlib</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בפרויקטים יהיה שימוש בגרפים. אז בואו נראה גרף אחד בצורה מלאה, בלי Pandas:<br>
נחשב את סכום המכירות לכל קטגוריה, נצייר גרף עמודות, ונשמור אותו לקובץ <code>sales_chart.png</code> כדי שנוכל לצרף אותו למייל.

```python
import matplotlib.pyplot as plt
import pandas as pd

sales = pd.read_csv('../../resources/sales.csv')

# סכום המכירות לכל קטגוריה
by_category = sales.groupby('category')['total'].sum()

# matplotlib לא מציג עברית נכון, לכן בגרף נחליף לשמות באנגלית
by_category.index = by_category.index.map({'אלקטרוניקה': 'Electronics', 'ריהוט': 'Furniture'})

plt.figure(figsize=(6, 4))
plt.bar(by_category.index, by_category.values, color=['#4C78A8', '#F58518'])
plt.title('Sales per category')
plt.ylabel('Total (NIS)')
plt.tight_layout()
plt.savefig('sales_chart.png')  # שמירה לקובץ - אפשר לצרף למייל
plt.show()

print("הגרף נשמר בקובץ sales_chart.png")
```

## <p style="text-align: right; direction: rtl; float: right; clear: both;">כלי עזר משותף: שמירת מייל לקובץ</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בכיתה לא שולחים מיילים אמיתיים. התא הבא מגדיר פונקציה אחת, <code>save_email_to_file</code>, שבונה מייל (עם קבצים מצורפים) ושומרת אותו ל-<code>.eml</code>.<br>
<strong>הריצו אותו לפני שמתחילים את הפרויקט.</strong> שליחה אמיתית אפשרית בבית בלבד, עם הקוד מיום 7.
</p>

```python
import os
from email.mime.text import MIMEText
from email.mime.multipart import MIMEMultipart
from email.mime.base import MIMEBase
from email import encoders


def save_email_to_file(subject, body, filename, attachments=()):
    """בונה מייל (עם קבצים מצורפים אם יש) ושומרת אותו לקובץ eml - בלי לשלוח!"""
    msg = MIMEMultipart()
    msg['From'] = 'reports@example.com'
    msg['To'] = 'manager@example.com'
    msg['Subject'] = subject
    msg.attach(MIMEText(body, 'plain', 'utf-8'))

    for attachment_path in attachments:
        if os.path.exists(attachment_path):
            with open(attachment_path, 'rb') as f:
                part = MIMEBase('application', 'octet-stream')
                part.set_payload(f.read())
            encoders.encode_base64(part)
            part.add_header('Content-Disposition', 'attachment',
                           filename=('utf-8', '', os.path.basename(attachment_path)))
            msg.attach(part)

    with open(filename, 'wb') as f:
        f.write(msg.as_bytes())
    print(f"המייל נשמר ב-{filename} (אפשר לפתוח אותו בתוכנת המייל)")


# בדיקה קטנה
save_email_to_file('בדיקה', 'שלום! זה מייל בדיקה.', 'test_mail.eml')
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>תא A - טעינת נתונים ו-API:</strong>

```python
import os
import sqlite3
import pandas as pd
import requests

if os.path.exists('project2_prices.db'):
    os.remove('project2_prices.db')

# דרישה 1 - היסטוריית מחירים: date, product, price (14 ימים, 3 מוצרים)
prices = pd.read_csv('../../resources/price_history.csv')
print(prices.head())

conn = sqlite3.connect('project2_prices.db')
prices.to_sql('prices', conn, index=False, if_exists='replace')


def get_current_price(product_id=1):
    """שולפת מחיר חי מ-API פתוח (dummyjson). אם אין אינטרנט - מחזירה None"""
    try:
        response = requests.get(f"https://dummyjson.com/products/{product_id}", timeout=10)
        response.raise_for_status()
        data = response.json()
        return data['title'], data['price']
    except Exception as e:
        print("לא הצלחנו להתחבר ל-API:", e)
        return None


print(get_current_price())
# TODO (דרישה 2): אם get_current_price החזירה תוצאה - הוסיפו אותה כשורה חדשה בטבלה prices עם INSERT (התאריך של היום)
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>תא B - ניתוח:</strong>
</p>

```python
# דרישה 3: סטטיסטיקות לכל מוצר - רק ההיסטוריה מהקובץ (עד יוני 2024), בלי שורת המחיר החי
stats = pd.read_sql(
    "SELECT product, MIN(price) AS min_price, MAX(price) AS max_price, ROUND(AVG(price), 1) AS avg_price "
    "FROM prices WHERE date < '2024-07-01' GROUP BY product", conn)
print(stats)

# שורה חדשה (לא למדנו אותה, אבל הנה הסבר):
# pct_change() מחשבת את אחוז השינוי לעומת השורה הקודמת. groupby מבטיח שזה יקרה בנפרד לכל מוצר.
history = pd.read_sql("SELECT * FROM prices WHERE date < '2024-07-01' ORDER BY product, date", conn)
history['change_pct'] = history.groupby('product')['price'].pct_change() * 100

# דרישה 4: ירידות של יותר מ-5% לעומת היום הקודם של אותו מוצר
drops = history[history['change_pct'] < -5]
print(drops)

# TODO (בונוס): חשבו ממוצע נע של 3 ימים (רמז: .groupby('product')['price'].rolling(3).mean())
```

```python
# דרישה 5 - דוח Excel
with pd.ExcelWriter('price_report.xlsx') as writer:
    stats.to_excel(writer, sheet_name='stats', index=False)
    drops.to_excel(writer, sheet_name='drops', index=False)

print("הדוח נשמר ב-price_report.xlsx")

# TODO (דרישה 6): אם יש ירידות - כתבו מייל התראה ושמרו אותו ל-alert.eml עם save_email_to_file (מצרפים את price_report.xlsx)
# TODO (דרישה 7): גרף קו של מחיר לאורך זמן, קו לכל מוצר, שמרו ב-price_chart.png (כותרות באנגלית)

conn.close()
```

    <li>&#9744; הקובץ <code>emails_report.xlsx</code> נפתח ויש בו הגיליונות <code>all_emails</code> ו-<code>summary</code></li>
    <li>&#9744; בחרו 3 מיילים אקראיים ובדקו ידנית שהקטגוריה שלהם מתאימה לכללים בדרישה 2</li>
    <li>&#9744; הקובץ <code>daily_summary.eml</code> נפתח בתוכנת מייל, מופיעים בו הנתונים מדרישה 7 והקובץ המצורף</li>

```python
import os
import sqlite3
import pandas as pd

if os.path.exists('project3_emails.db'):
    os.remove('project3_emails.db')

emails = pd.read_csv('../../resources/emails.csv')
print(emails[['id', 'sender', 'subject', 'attachment']].head())


def classify_email(subject, body):
    """מחזירה קטגוריה לפי מילות מפתח: חשבונית / הזמנה / משימה / אחר"""
    # רמז: מייל אחד יכול להכיל כמה מילות מפתח (למשל הזמנה שמבקשת חשבונית).
    # לכן בודקים קודם את הנושא, ורק אם לא מצאנו - את התוכן.
    if 'חשבונית' in subject:
        return 'חשבונית'
    # TODO (דרישה 2): הוסיפו תנאים להזמנה, למשימה ולפרסומת ("מבצע", "זכית", "ניוזלטר") לפי הסדר שבדרישה, ובדיקת תוכן המייל לחשבונית
    return 'אחר'


emails['category'] = [classify_email(subject, body) for subject, body in zip(emails['subject'], emails['body'])]
print(emails['category'].value_counts())

conn = sqlite3.connect('project3_emails.db')
emails.to_sql('emails', conn, index=False, if_exists='replace')
```
