## <p style="text-align: right; direction: rtl; float: right; clear: both;">כלי עזר משותף: שמירת מייל לקובץ</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בכיתה לא שולחים מיילים אמיתיים. התא הבא מגדיר פונקציה אחת, <code>save_email_to_file</code>, שבונה מייל (עם קבצים מצורפים) ושומרת אותו ל-<code>.eml</code>.<br>
<strong>הריצו אותו לפני שמתחילים את הפרויקט.</strong> שליחה אמיתית אפשרית בבית בלבד, עם הקוד מיום 7.
</p>

```python
import os
from email.mime.text import MIMEText
from email.mime.multipart import MIMEMultipart
from email.mime.base import MIMEBase
from email import encoders


def save_email_to_file(subject, body, filename, attachments=()):
    """בונה מייל (עם קבצים מצורפים אם יש) ושומרת אותו לקובץ eml - בלי לשלוח!"""
    msg = MIMEMultipart()
    msg['From'] = 'reports@example.com'
    msg['To'] = 'manager@example.com'
    msg['Subject'] = subject
    msg.attach(MIMEText(body, 'plain', 'utf-8'))

    for attachment_path in attachments:
        if os.path.exists(attachment_path):
            with open(attachment_path, 'rb') as f:
                part = MIMEBase('application', 'octet-stream')
                part.set_payload(f.read())
            encoders.encode_base64(part)
            part.add_header('Content-Disposition', 'attachment',
                           filename=('utf-8', '', os.path.basename(attachment_path)))
            msg.attach(part)

    with open(filename, 'wb') as f:
        f.write(msg.as_bytes())
    print(f"המייל נשמר ב-{filename} (אפשר לפתוח אותו בתוכנת המייל)")


# בדיקה קטנה
save_email_to_file('בדיקה', 'שלום! זה מייל בדיקה.', 'test_mail.eml')
```

## <p style="text-align: right; direction: rtl; float: right; clear: both;">כלי עזר משותף: בדיקות עצמיות</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בכל פרויקט יש תא <strong>בדיקות עצמיות</strong>. הוא בודק את הקוד שלכם מול הדרישות ומדפיס שורה לכל בדיקה. את התא הבא מריצים פעם אחת, לפני הפרויקט: הוא מגדיר את הפונקציה <code>check(description, func)</code> שהבדיקות משתמשות בה, ועוד שני עוזרים קטנים (<code>raises</code> ו-<code>read_eml</code>).<br>
<code>check</code> מקבלת תיאור של הבדיקה ופונקציה קטנה (<code>lambda</code>) שמחזירה <code>True</code> אם הכל תקין. היא מריצה את הפונקציה בעצמה, ולכן שגיאה בקוד שלכם <strong>לא מפילה את המחברת</strong> - היא רק מופיעה בשורה של הבדיקה.
</p>

```python
import os
import tempfile
from email import policy
from email import message_from_binary_file

CHECK_HINTS = {
    'NameError': 'משהו עדיין לא הוגדר - הריצו את התא שמגדיר אותו',
    'AttributeError': 'חסרה תכונה או מתודה',
    'TypeError': 'הקריאה לא מתאימה להגדרה (או שהקוד עדיין לא נכתב)',
    'ValueError': 'התקבל ערך לא תקין',
    'KeyError': 'חסר מפתח',
    'FileNotFoundError': 'קובץ לא נמצא',
}


def check(description, func):
    """מריצה בדיקה אחת. func צריכה להחזיר True. מדפיסה ✅ או ❌ - בלי traceback ארוך"""
    try:
        passed = bool(func())
    except Exception as error:
        hint = CHECK_HINTS.get(type(error).__name__, 'התרחשה שגיאה')
        print(f"❌ {description}  |  {hint}: {error}")
        return False
    if passed:
        print(f"✅ {description}")
    else:
        print(f"❌ {description}  |  התוצאה לא תואמת את הדרישה")
    return passed


def raises(error_type, func):
    """מחזירה True אם הרצת func גרמה לשגיאה מהסוג error_type (למשל ValueError)"""
    try:
        func()
    except error_type:
        return True
    return False


def read_eml(path):
    """קוראת קובץ eml. מחזירה (גוף המייל, רשימת שמות הקבצים המצורפים)"""
    with open(path, 'rb') as f:
        msg = message_from_binary_file(f, policy=policy.default)
    body = msg.get_body(preferencelist=('plain',)).get_content()
    return body, [part.get_filename() for part in msg.iter_attachments()]
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
<code>check</code> מריצה את <code>func()</code> בתוך <code>try/except</code>. אם התוצאה <code>True</code> - מדפיסה ✅. אם התוצאה לא <code>True</code>, או שהייתה שגיאה - מדפיסה ❌, ואחרי הקו האנכי <code>|</code> הסבר קצר בעברית. <code>raises(ValueError, func)</code> מחזירה <code>True</code> אם <code>func</code> זרקה <code>ValueError</code> - כך בודקים שהולידציה עובדת.
</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בתא הבא יש דוגמה לארבע בדיקות. <strong>שתי שורות ה-❌ כאן הן בכוונה</strong>, כדי שתראו איך נראית בדיקה שנכשלת ובדיקה שקורסת:
</p>

```python
check("2 ועוד 2 הם 4", lambda: 2 + 2 == 4)
check("בדיקה שנכשלת: 2 ועוד 2 הם 5", lambda: 2 + 2 == 5)
check("בדיקה שקורסת: חלוקה באפס", lambda: 1 / 0 == 0)
check("int('abc') זורקת ValueError", lambda: raises(ValueError, lambda: int('abc')))
```

### <p style="text-align: right; direction: rtl; float: right; clear: both;">איך קוראים את הבדיקות</p>

<ul style="text-align: right; direction: rtl; float: right; clear: both;">
    <li><strong>✅</strong> - הדרישה מתקיימת</li>
    <li><strong>❌ והתוצאה לא תואמת</strong> - הקוד רץ, אבל מחזיר משהו אחר ממה שהדרישה מבקשת. קראו שוב את הדרישה (המספר כתוב בתחילת השורה)</li>
    <li><strong>❌ וסוג שגיאה</strong> (למשל "חסרה תכונה או מתודה") - הקוד קרס. ברוב המקרים זה אומר שעוד לא כתבתם את החלק הזה, או ששם התכונה או המתודה שונה ממה שבדרישה</li>
    <li>בתחילת העבודה רוב השורות יהיו ❌ - זה תקין. עובדים שורה אחרי שורה עד שהכל ✅</li>
    <li><strong>אחרי שינוי במחלקה:</strong> הריצו מחדש את התא שמגדיר אותה, ורק אז את תא הבדיקות</li>
</ul>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">כלי עזר משותף: מחלקת Report (בסיס לכל הדוחות)</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בכל הפרויקטים יש דוחות: דוח Excel, גרף, מייל. לכולם יש כותרת, כולם "בונים" תוכן, וכולם נשמרים לקובץ. במקום לכתוב את החלקים המשותפים שוב בכל דוח, כותבים אותם פעם אחת במחלקת בסיס, <code>Report</code>, וכל דוח <strong>יורש</strong> ממנה (ירושה, כמו שלמדנו ביום 2).<br>
<strong>הריצו את התא הבא לפני שמתחילים את הפרויקט.</strong> שימו לב ל-<code>build()</code>: ב-<code>Report</code> היא רק זורקת <code>NotImplementedError</code>, כי רק מחלקה יורשת יודעת מה להכניס לדוח שלה.
</p>

```python
class Report:
    """דוח כללי: יש לו כותרת, מתודה build() שבונה את התוכן, ומתודות ששומרות אותו"""

    def __init__(self, title):
        self.title = title

    def build(self):
        """כל סוג דוח חייב לממש את build() בעצמו"""
        raise NotImplementedError("צריך לממש את build() במחלקה היורשת")

    def save(self, path):
        """שומרת את תוכן הדוח (טקסט) לקובץ ומחזירה את הנתיב"""
        content = self.build()
        with open(path, 'w', encoding='utf-8') as f:
            f.write(content)
        return path

    def save_as_email(self, path, attachments=()):
        """שומרת את הדוח כקובץ eml: הכותרת היא נושא המייל, ו-build() הוא הגוף"""
        save_email_to_file(self.title, self.build(), path, attachments)
        return path

    def __str__(self):
        return f"{type(self).__name__}: {self.title}"
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong>
</p>

<ul style="text-align: right; direction: rtl; float: right; clear: both;">
    <li><code>__init__</code> שומרת את הכותרת, ו-<code>__str__</code> מציגה את שם המחלקה (<code>type(self).__name__</code>) ואת הכותרת</li>
    <li><code>build()</code> זורקת <code>NotImplementedError</code> - "חוזה" שמחלקה יורשת חייבת לממש</li>
    <li><code>save(path)</code> מסתמכת על <code>build()</code> (שנכתבת במחלקה היורשת) ושומרת את הטקסט לקובץ. מחלקה יורשת יכולה לדרוס אותה אם היא שומרת משהו אחר, למשל קובץ Excel או תמונה</li>
    <li><code>save_as_email</code> משתמשת בפונקציה <code>save_email_to_file</code> מהחלק הקודם: הכותרת היא נושא המייל, ותוצאת <code>build()</code> היא גוף המייל</li>
</ul>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
דוגמה מלאה לדוח שיורש מ-<code>Report</code>. בדיוק את אותו הדפוס תכתבו בפרויקט: בנאי שקורא ל-<code>super().__init__</code>, ו-<code>build()</code>. כל השאר מגיע בירושה:
</p>

```python
class GreetingReport(Report):
    """דוגמה לדוח שיורש מ-Report: רק build() כתובה כאן, כל השאר מגיע מהמחלקה Report"""

    def __init__(self, name):
        super().__init__("ברכה")  # מפעילים את הבנאי של Report כדי לשמור את הכותרת
        self.name = name

    def build(self):
        return f"שלום {self.name}, זה דוח הדגמה."


greeting = GreetingReport("דנה")
print(greeting)                               # __str__ מגיעה מ-Report
print(greeting.build())
greeting.save("demo_report.txt")              # save מגיעה מ-Report
greeting.save_as_email("demo_report.eml")     # גם save_as_email מגיעה מ-Report

try:
    Report("דוח ריק").build()                 # ל-Report עצמה אין build אמיתית
except NotImplementedError as error:
    print("כצפוי:", error)
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
ב-<code>GreetingReport</code> כתבנו רק את <code>__init__</code> ואת <code>build</code>, ובכל זאת הדפסה (<code>__str__</code>), שמירה לקובץ (<code>save</code>) ושמירה כמייל (<code>save_as_email</code>) עובדות - הן מגיעות מ-<code>Report</code>. <code>super().__init__("ברכה")</code> מפעילה את הבנאי של <code>Report</code>. ואילו <code>Report("דוח ריק").build()</code> נכשלת בכוונה עם <code>NotImplementedError</code>, ואנחנו תופסים את זה עם <code>try/except</code>.
</p>

```python

class Product:
    """מוצר והיסטוריית המחירים שלו (דרישות 1-4 ו-6)."""

    def __init__(self, name):
        # TODO (דרישות 1, 2): בדקו את השם ושמרו name ו-history
        if not name.strip():
            raise ValueError("שם המוצר לא יכול להיות ריק")
        self.name = name
        self.history = []

    def add_price(self, date, price):
        # TODO (דרישות 1, 2): בדקו מחיר והוסיפו לרשימה ממוינת
        if price <= 0:
            raise ValueError("המחיר חייב להיות חיובי")
        self.history.append((date, price))
        self.history.sort()

    # TODO (דרישה 1): __len__
    def __len__(self):
        return len(self.history)

    # TODO (דרישה 3): ארבע properties
    @property
    def min_price(self):
        return min(price for date, price in self.history) if self.history else None

    @property
    def max_price(self):
        return max(price for date, price in self.history) if self.history else None

    @property
    def average_price(self):
        return round(sum(price for date, price in self.history) / len(self), 1) if self.history else None

    @property
    def latest_price(self):
        return self.history[-1][1] if self.history else None

    # TODO (דרישה 4): __str__
    def __str__(self):
        return (f"{self.name}: {len(self)} מחירים "
                f"(מינימום {self.min_price}, מקסימום {self.max_price}, "
                f"ממוצע {self.average_price}, אחרון {self.latest_price})")

    def find_drops(self):
        # TODO (דרישה 6): מציאת ירידות מחיר
        drops = []
        for (old_date, old_price), (date, new_price) in zip(self.history, self.history[1:]):
            drop = PriceDrop(self.name, date, old_price, new_price)
            if drop.change_pct < PriceDrop.THRESHOLD:
                drops.append(drop)
        return drops


class PriceDrop:
    """ירידת מחיר אחת (דרישה 5)."""

    # TODO (דרישה 5): THRESHOLD
    THRESHOLD = -5

    def __init__(self, product_name, date, old_price, new_price):
        # TODO (דרישה 5): שמירת התכונות
        self.product_name = product_name
        self.date = date
        self.old_price = old_price
        self.new_price = new_price

    # TODO (דרישה 5): change_pct, __str__, __lt__
    @property
    def change_pct(self):
        return (self.new_price - self.old_price) / self.old_price * 100

    def __str__(self):
        return f"{self.date}: {self.product_name} ירד ב-{abs(self.change_pct):.1f}% (למחיר {self.new_price})"

    def __lt__(self, other):
        return self.change_pct < other.change_pct
```

```python

import csv
import datetime
import os
import sqlite3
import pandas as pd
import requests


def get_current_price(product_id=1):
    """שולפת מחיר חי מ-API פתוח (dummyjson). אם אין אינטרנט - מחזירה None"""
    try:
        response = requests.get(f"https://dummyjson.com/products/{product_id}", timeout=10)
        response.raise_for_status()
        data = response.json()
        return data['title'], data['price']
    except Exception as e:
        print("לא הצלחנו להתחבר ל-API:", e)
        return None


class PriceTracker:
    """מסד המחירים והניתוח שלהם (דרישות 7-14)."""

    HISTORY_END = '2024-06-30'

    def __init__(self, db_path):
        # TODO (דרישה 7): חיבור למסד ויצירת טבלה
        self.conn = sqlite3.connect(db_path)
        self.conn.execute(
            "CREATE TABLE IF NOT EXISTS prices "
            "(date TEXT, product TEXT, price NUMERIC)"
        )
        self.conn.commit()

    def add_price(self, date, product, price):
        # TODO (דרישה 7): בדיקת הערכים והוספה לטבלה
        Product(product).add_price(date, price)
        self.conn.execute(
            "INSERT INTO prices (date, product, price) VALUES (?, ?, ?)",
            (date, product, price)
        )
        self.conn.commit()

    def __len__(self):
        # TODO (דרישה 7): מספר השורות
        return self.conn.execute(
            "SELECT COUNT(*) FROM prices"
        ).fetchone()[0]

    def close(self):
        # TODO (דרישה 7): סגירת החיבור
        self.conn.close()

    def load_csv(self, csv_path):
        # TODO (דרישה 8): טעינת CSV
        if not os.path.exists(csv_path):
            print("קובץ המחירים לא נמצא:", csv_path)
            return 0

        count = 0
        with open(csv_path, encoding='utf-8-sig', newline='') as file:
            for row in csv.DictReader(file):
                try:
                    self.add_price(
                        row['date'],
                        row['product'],
                        int(row['price'])
                    )
                    count += 1
                except ValueError as error:
                    print("דילוג על שורה לא תקינה:", error)

        return count

    def get_products(self):
        """דרישה 9 (דוגמה מוכנה): רשימת Product מההיסטוריה בטבלה"""
        rows = self.conn.execute(
            "SELECT date, product, price FROM prices "
            "WHERE date <= ? ORDER BY product, date",
            (self.HISTORY_END,)
        ).fetchall()

        products = {}
        for date, name, price in rows:
            if name not in products:
                products[name] = Product(name)
            products[name].add_price(date, price)

        return list(products.values())

    def fetch_live_price(self, product_id=1, fetcher=get_current_price):
        # TODO (דרישה 10): שליפת מחיר חי
        result = fetcher(product_id)

        if result is None:
            return None

        title, price = result
        self.add_price(datetime.date.today().isoformat(), title, price)

        return title, price

    # ---------- ניתוח ----------

    def stats(self):
        # TODO (דרישה 11): טבלת סטטיסטיקות
        rows = []
        for product in self.get_products():
            rows.append({
                'product': product.name,
                'min_price': product.min_price,
                'max_price': product.max_price,
                'avg_price': product.average_price
            })

        return pd.DataFrame(
            rows,
            columns=['product', 'min_price', 'max_price', 'avg_price']
        )

    def all_drops(self):
        # TODO (דרישה 12): כל ירידות המחיר
        drops = []
        for product in self.get_products():
            drops.extend(product.find_drops())

        return sorted(drops)

    def drops_table(self):
        # TODO (דרישה 13): טבלת ירידות
        rows = []
        for drop in self.all_drops():
            rows.append({
                'date': drop.date,
                'product': drop.product_name,
                'price': drop.new_price,
                'change_pct': drop.change_pct
            })

        return pd.DataFrame(
            rows,
            columns=['date', 'product', 'price', 'change_pct']
        )

    def biggest_drop(self):
        # TODO (דרישה 14): הירידה הגדולה ביותר
        drops = self.all_drops()
        return min(drops) if drops else None

    # TODO (בונוס): ב-Product - מתודה moving_average(window=3)
```

```python

import matplotlib.pyplot as plt

# matplotlib לא מציג עברית נכון, לכן בגרף משתמשים בשמות באנגלית
ENGLISH_NAMES = {
    'אוזניות אלחוטיות': 'Wireless headphones',
    'מסך 24 אינץ': 'Monitor 24',
    'מקלדת מכנית': 'Mechanical keyboard'
}


class PriceExcelReport(Report):
    """דוח Excel (דרישה 15). יורשת מ-Report"""

    def __init__(self, tracker):
        # TODO (דרישה 15): super ושמירת tracker
        super().__init__('דוח מחירים')
        self.tracker = tracker

    def build(self):
        # TODO (דרישה 15): שני גיליונות
        return {
            'stats': self.tracker.stats(),
            'drops': self.tracker.drops_table()
        }

    def save(self, path):
        """כותבת את הגיליונות לקובץ Excel ומחזירה את path"""
        with pd.ExcelWriter(path) as writer:
            # TODO (דרישה 15): לולאה על כל הגיליונות
            for sheet_name, table in self.build().items():
                table.to_excel(
                    writer,
                    sheet_name=sheet_name,
                    index=False
                )
        return path

    # TODO (דרישה 15): __str__
    def __str__(self):
        return super().__str__() + ' (2 גיליונות)'


class PriceChartReport(Report):
    """גרף קו של המחירים (דרישה 16). יורשת מ-Report"""

    def __init__(self, tracker):
        # TODO (דרישה 16): super ושמירת tracker
        super().__init__('גרף מחירים')
        self.tracker = tracker

    def build(self):
        # TODO (דרישה 16): קו לכל מוצר
        fig, ax = plt.subplots(figsize=(9, 5))

        for product in self.tracker.get_products():
            dates = [date for date, price in product.history]
            prices = [price for date, price in product.history]

            ax.plot(
                dates,
                prices,
                label=ENGLISH_NAMES[product.name]
            )

        ax.set_title('Product prices')
        ax.set_xlabel('Date')
        ax.set_ylabel('Price')
        ax.legend()

        fig.autofmt_xdate()
        fig.tight_layout()

        return fig

    def save(self, path):
        # TODO (דרישה 16): שמירת PNG וסגירת הגרף
        fig = self.build()
        fig.savefig(path)
        plt.close(fig)
        return path


class PriceAlertReport(Report):
    """מייל התראה על ירידות (דרישה 17). יורשת מ-Report"""

    def __init__(self, tracker):
        # TODO (דרישה 17): super ושמירת tracker
        super().__init__('התראת ירידת מחיר')
        self.tracker = tracker

    # TODO (דרישה 17): property בשם has_alerts
    @property
    def has_alerts(self):
        return bool(self.tracker.all_drops())

    def build(self):
        # TODO (דרישה 17): טקסט ההתראה
        drops = self.tracker.all_drops()

        if not drops:
            return 'לא זוהו ירידות מחיר'

        return 'זוהו ירידות מחיר:\n' + '\n'.join(
            str(drop) for drop in drops
        )
```

```python

class PriceProject:
    """מחברת הכל (דרישה 18): PriceProject(csv_path, db_path).run()"""

    def __init__(self, csv_path, db_path):
        # TODO (דרישה 18): שמירת הנתיבים
        self.csv_path = csv_path
        self.db_path = db_path

    def run(self):
        """מריצה את כל הפרויקט ומחזירה את רשימת הקבצים שנוצרו"""

        # TODO (דרישה 18): בדיקת קובץ ה-CSV
        if not os.path.exists(self.csv_path):
            print("קובץ המחירים לא נמצא:", self.csv_path)
            return []

        # מחיקת מסד ישן
        if os.path.exists(self.db_path):
            os.remove(self.db_path)

        tracker = PriceTracker(self.db_path)
        created_files = []

        try:
            # טעינת הנתונים
            tracker.load_csv(self.csv_path)

            # שליפת מחיר חי
            live_price = tracker.fetch_live_price()
            print(
                "מחיר חי:",
                live_price if live_price is not None else "לא זמין"
            )

            # שמירת דוח Excel
            created_files.append(
                PriceExcelReport(tracker).save('price_report.xlsx')
            )

            # שמירת התראה רק אם נמצאו ירידות
            alert = PriceAlertReport(tracker)
            if alert.has_alerts:
                created_files.append(
                    alert.save_as_email(
                        'alert.eml',
                        ['price_report.xlsx']
                    )
                )

            # שמירת גרף
            created_files.append(
                PriceChartReport(tracker).save('price_chart.png')
            )

            return created_files

        finally:
            # סגירת מסד הנתונים גם במקרה של שגיאה
            tracker.close()
```
