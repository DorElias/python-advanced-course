## <p style="text-align: right; direction: rtl; float: right; clear: both;">יצירת DataFrame ראשון</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>DataFrame</strong> זה כמו טבלת Excel - יש לו שורות ועמודות.
נתחיל ביצירה פשוטה מתוך מילון:
</p>

```python
import pandas as pd

# יצירת DataFrame מתוך מילון
# כל מפתח במילון יהפוך לשם עמודה
נתונים = {
    'שם': ['דני', 'שרה', 'יוסי', 'מיכל'],
    'גיל': [25, 30, 28, 35],
    'עיר': ['תל אביב', 'חיפה', 'ירושלים', 'באר שבע'],
    'משכורת': [10000, 15000, 12000, 18000]
}

df = pd.DataFrame(נתונים)
print(df)
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
יצרנו מילון Python רגיל, שבו כל מפתח מייצג עמודה וכל רשימה מכילה את הערכים באותה עמודה.<br>
Pandas אוטומטית יצר אינדקס (מספור השורות) - 0, 1, 2, 3.<br>
התוצאה היא טבלה מסודרת ונוחה לקריאה!
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">קריאת נתונים מ-CSV</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>CSV</strong> (Comma Separated Values) זה פורמט קובץ נפוץ מאוד לשמירת טבלאות.<br>
Excel יכול לפתוח CSV, ורוב המערכות יודעות לייצא אליו נתונים.<br>
Pandas הופך קריאת CSV לפעולה של שורה אחת!
</p>

```python
# קריאת קובץ CSV
# הנתיב הוא יחסי למיקום המחברת
df_עובדים = pd.read_csv('../../resources/employees.csv')

# הצגת 5 שורות ראשונות
print("5 שורות ראשונות:")
print(df_עובדים.head())

print("\n" + "="*50 + "\n")

# הצגת 3 שורות אחרונות
print("3 שורות אחרונות:")
print(df_עובדים.tail(3))
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>למה להשתמש ב-head() ו-tail()?</strong><br>
כשיש לנו מאות או אלפי שורות, אנחנו לא רוצים להדפיס את הכל!<br>
<code>head()</code> מציג את תחילת הטבלה - מעולה לבדוק שהקובץ נטען נכון.<br>
<code>tail()</code> מציג את סוף הטבלה - שימושי לוודא שכל הנתונים נטענו.
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">מידע על ה-DataFrame</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
לפני שמתחילים לעבוד עם נתונים, חשוב להבין מה יש לנו בטבלה.<br>
Pandas מספק כלים נהדרים לכך:
</p>

```python
# מידע כללי על הטבלה
print("מידע על הטבלה:")
df_עובדים.info()  # info מדפיסה בעצמה - לא עוטפים ב-print

print("\n" + "="*50 + "\n")

# גודל הטבלה (שורות, עמודות)
print(f"גודל הטבלה: {df_עובדים.shape}")
print(f"מספר שורות: {df_עובדים.shape[0]}")
print(f"מספר עמודות: {df_עובדים.shape[1]}")

print("\n" + "="*50 + "\n")

# שמות העמודות
print("שמות העמודות:")
print(df_עובדים.columns.tolist())
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה למדנו?</strong><br>
<code>info()</code> - מדפיסה בעצמה (לכן לא כותבים <code>print(df.info())</code> - זה היה מדפיס גם את המילה None) את סוג הנתונים בכל עמודה, האם יש ערכים חסרים, וכמה זיכרון הטבלה תופסת.<br>
<code>shape</code> - נותן טאפל (tuple) עם מספר השורות והעמודות.<br>
<code>columns</code> - נותן גישה לשמות כל העמודות.
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">סטטיסטיקות בסיסיות</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
אחת הפונקציות החזקות של Pandas היא <code>describe()</code>.<br>
היא מחשבת אוטומטית סטטיסטיקות עבור כל העמודות המספריות!
</p>

```python
# סטטיסטיקות מפורטות
print(df_עובדים.describe())
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה רואים כאן?</strong>
</p>

<ul style="text-align: right; direction: rtl; float: right; clear: both;">
    <li><strong>count</strong> - כמה ערכים יש (לא כולל חסרים)</li>
    <li><strong>mean</strong> - ממוצע</li>
    <li><strong>std</strong> - סטיית תקן (מידת הפיזור של הנתונים)</li>
    <li><strong>min/max</strong> - ערך מינימלי ומקסימלי</li>
    <li><strong>25%, 50%, 75%</strong> - אחוזונים (percentiles)</li>
</ul>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
מדד 50% זה החציון (median) - הערך האמצעי כשממיינים את כל הנתונים.
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">גישה לעמודות</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
יש 2 דרכים עיקריות לגשת לעמודה:
</p>

```python
# דרך 1: סוגריים מרובעים (כמו מילון)
print("גישה דרך סוגריים:")
print(df['שם'])

print("\n" + "="*50 + "\n")

# דרך 2: נקודה - עובדת כאשר שם העמודה הוא שם משתנה "חוקי" (בלי רווחים)
# זה עובד גם בעברית!
print("גישה דרך נקודה:")
print(df.שם)
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>איזו דרך עדיפה?</strong><br>
סוגריים מרובעים (<code>df['עמודה']</code>) - עובדים תמיד, גם עם רווחים וגם עם שמות מיוחדים.<br>
נקודה (<code>df.עמודה</code>) - נוחה לקריאה ועובדת גם בעברית, אבל <strong>לא מומלצת</strong> כשהשם כולל רווחים,
או כשהוא זהה לשם של מתודה קיימת (למשל עמודה בשם <code>count</code> או <code>size</code>).<br>
בנוסף, אי אפשר ליצור עמודה חדשה עם נקודה - רק עם סוגריים.
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">בחירת כמה עמודות</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
לפעמים נרצה לראות רק חלק מהעמודות. נעביר רשימה של שמות:
</p>

```python
# בחירת עמודות ספציפיות
עמודות_נבחרות = ['שם', 'עיר']
print(df[עמודות_נבחרות])
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>למה זה שימושי?</strong><br>
כשיש 50 עמודות אבל אנחנו מתעניינים רק ב-3, קל יותר לראות את מה שחשוב.<br>
זה גם חוסך זיכרון אם נשמור את ה-DataFrame המצומצם.
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">גישה לשורות - loc ו-iloc</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
ב-Pandas יש 2 דרכים עיקריות לגשת לשורות:
</p>

<ul style="text-align: right; direction: rtl; float: right; clear: both;">
    <li><strong>loc</strong> - גישה לפי שם האינדקס (label-based)</li>
    <li><strong>iloc</strong> - גישה לפי מיקום (position-based, כמו רשימה רגילה)</li>
</ul>

```python
# iloc - גישה לפי מיקום (מתחיל מ-0)
print("השורה הראשונה (מיקום 0):")
print(df.iloc[0])

print("\n" + "="*50 + "\n")

# iloc עם טווח
print("שורות 1-3 (מיקומים 1,2,3):")
print(df.iloc[1:4])  # שימו לב: 4 לא כלול!

print("\n" + "="*50 + "\n")

# loc - גישה לפי שם (כרגע האינדקס הוא 0,1,2...)
print("שורה באינדקס 0:")
print(df.loc[0])
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מתי להשתמש במה?</strong><br>
<code>iloc</code> - כשאתם חושבים במונחים של "תן לי שורה מספר 5" או "שורות 10-20".<br>
<code>loc</code> - כשיש לכם אינדקס משמעותי (למשל תאריכים או קוד עובד) ואתם רוצים לגשת לפי הערך.
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">שמירת DataFrame ל-CSV</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">קריאה וכתיבה של קבצי Excel</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בעבודה אמיתית הרבה נתונים מגיעים בקבצי Excel.<br>
Pandas קורא ושומר Excel בשורה אחת, בעזרת הספרייה <code>openpyxl</code> שכבר התקנו ביום 3 (<code>pip install -r requirements.txt</code>).<br>
בתיקיית <code>resources</code> יש קובץ <code>products.xlsx</code> עם רשימת מוצרים. נקרא אותו:
</p>

```python
# קריאת קובץ Excel - read_excel
df_מוצרים = pd.read_excel('../../resources/products.xlsx')
print(df_מוצרים)
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
<code>read_excel</code> עובד בדיוק כמו <code>read_csv</code>: מחזיר DataFrame מוכן.<br>
כברירת מחדל נקרא הגיליון הראשון. אפשר לבחור גיליון אחר עם <code>sheet_name='שם הגיליון'</code>.
</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
עכשיו נוסיף עמודה מחושבת (שווי המלאי של כל מוצר) ונשמור דוח חדש ל-Excel:
</p>

```python
# עמודה חדשה: מחיר כפול מלאי
df_מוצרים['שווי_מלאי'] = df_מוצרים['מחיר'] * df_מוצרים['מלאי']

# שמירה לקובץ Excel חדש - to_excel
# אם הקובץ כבר קיים הוא יידרס, ולכן אפשר להריץ שוב ושוב
df_מוצרים.to_excel('inventory_report.xlsx', index=False, sheet_name='מלאי')
print("הקובץ inventory_report.xlsx נשמר!")

# בדיקה: קוראים אותו בחזרה
print(pd.read_excel('inventory_report.xlsx').head(3))
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
<code>to_excel</code> שומר DataFrame לקובץ <code>.xlsx</code>. כמו ב-CSV, <code>index=False</code> מונע שמירת עמודת המספור.<br>
את הקובץ אפשר לפתוח ב-Excel ולשלוח למנהל - בלי לכתוב אף שורה של openpyxl.
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">חיבור ל-SQL - שיעור 5 פוגש את Pandas</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
ביום 5 למדנו SQLite. Pandas יודע לקרוא תוצאה של שאילתת SQL ישר לתוך DataFrame עם <code>read_sql</code>,
ולשמור DataFrame כטבלה במסד נתונים עם <code>to_sql</code>.<br>
זה שילוב חזק: SQL מסנן ומקבץ במסד, ו-Pandas מנתח ומציג.
</p>

```python
import sqlite3
import os

# מתחילים נקי: אם המסד נשאר מהרצה קודמת - מוחקים אותו
if os.path.exists('employees_day6.db'):
    os.remove('employees_day6.db')

conn = sqlite3.connect('employees_day6.db')

# DataFrame -> טבלה במסד (to_sql)
df_עובדים.to_sql('employees', conn, index=False, if_exists='replace')

# שאילתת SQL -> DataFrame (read_sql)
שאילתה = """
SELECT department, COUNT(*) AS workers, AVG(salary) AS avg_salary
FROM employees
GROUP BY department
ORDER BY avg_salary DESC
"""
df_מחלקות = pd.read_sql(שאילתה, conn)
print(df_מחלקות)

conn.close()
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
<code>to_sql</code> יצר טבלה בשם <code>employees</code> מתוך ה-DataFrame (<code>if_exists='replace'</code> = אם הטבלה קיימת, מחליפים אותה).<br>
<code>read_sql</code> הריץ שאילתת <code>GROUP BY</code> מיום 5 והחזיר את התוצאה כ-DataFrame.<br>
בסוף סוגרים את החיבור עם <code>conn.close()</code>.
</p>
