## <p style="text-align: right; direction: rtl; float: right; clear: both;">למה צריך ניהול גרסאות?</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בטח נתקלתם בזה: כותבים סקריפט שעובד, ואז רוצים לשנות בו משהו. מתוך פחד לקלקל, שומרים עותק.<br>
ואז עוד עותק. ואז עוד אחד. בסוף התיקייה נראית ככה:
</p>

<ul style="text-align: right; direction: rtl; float: right; clear: both;">
    <li><code>sales_report.py</code></li>
    <li><code>sales_report_v2.py</code></li>
    <li><code>sales_report_v2_good.py</code></li>
    <li><code>sales_report_FINAL.py</code></li>
    <li><code>sales_report_FINAL_really_final.py</code></li>
</ul>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
ועכשיו, איזה מהם הכי עדכני? מה בדיוק השתנה בין <code>v2</code> ל-<code>FINAL</code>? ומה עושים אם הגרסה שעבדה נמחקה?<br>
<strong>מערכת לניהול גרסאות</strong> (Version Control System) פותרת את כל זה. <strong>Git</strong> היא המערכת הפופולרית בעולם.
</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה git נותן לנו?</strong>
</p>

<ul style="text-align: right; direction: rtl; float: right; clear: both;">
    <li><strong>היסטוריה:</strong> כל שמירה (commit) נרשמת עם תיאור, תאריך ושם. תמיד אפשר לראות מה השתנה, מתי ולמה.</li>
    <li><strong>ניסויים בטוחים:</strong> אפשר לנסות רעיון בענף נפרד, ואם הוא לא עבד פשוט זורקים אותו.</li>
    <li><strong>חזרה אחורה:</strong> אפשר להחזיר קובץ בודד לצורתו לפני שבוע.</li>
    <li><strong>עבודה משותפת:</strong> כמה אנשים עובדים על אותו פרויקט, ו-git עוזר לחבר את העבודות.</li>
    <li><strong>גיבוי:</strong> כשמעלים עותק לשרת (כמו GitHub), יש עותק נוסף של הכול מחוץ למחשב.</li>
</ul>

יש גם שירותים דומים ל-GitHub (GitLab, Bitbucket), וכולם עובדים עם אותו git.

</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">

אזור ההכנה מאפשר לבחור בדיוק מה ייכנס לכל commit, וכך ההיסטוריה נשארת מסודרת וברורה.
</p>

### <p style="text-align: right; direction: rtl; float: right; clear: both;">מה זה HEAD?</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<code>HEAD</code> הוא פשוט "<strong>אתם נמצאים כאן</strong>" - מצביע על המקום הנוכחי בהיסטוריה.<br>
HEAD מצביע על הענף שבו אתם עובדים (בינתיים זה <code>main</code>), והענף מצביע על ה-commit האחרון שלו.<br>
כל פעם שעושים commit חדש, הענף (ו-HEAD איתו) זז קדימה ל-commit החדש.<br>
בינתיים מספיק לזכור: HEAD = "הגרסה האחרונה שנשמרה, במקום שבו אני עומד".<br>
נחזור לענפים בהמשך, ואז התמונה תתבהר עוד יותר.

</p>

```python
!git --version
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">

אם קיבלתם הודעה כמו "command not found" או "'git' is not recognized" - git לא מותקן (או שהמחשב לא מוצא אותו).

    <li>אם אין אפשרות להתקין: אפשר לקרוא את המחברת בלי להריץ אותה. כל הפלטים מוסברים בטקסט.</li>
</ol>

### <p style="text-align: right; direction: rtl; float: right; clear: both;">הגדרת זהות (פעם אחת בחיים)</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
כל commit נושא שם ואימייל של מי שיצר אותו. לכן git צריך לדעת מי אתם.<br>
את זה מגדירים <strong>פעם אחת</strong> במחשב, באמצעות <code>git config --global</code> (המילה <code>global</code> אומרת: לכל הפרויקטים במחשב).<br>
<strong>אנחנו לא נריץ את הפקודות האלה במחברת</strong>, כדי לא לשנות את ההגדרות שלכם בלי שתרצו. הן מופיעות כאן כהערות, ואפשר להריץ אותן בעצמכם בטרמינל:
</p>

```python
# הפקודות הבאות הן הערות. הריצו אותן בטרמינל (עם השם והאימייל שלכם) אם אתם רוצים להגדיר זהות.
# git config --global user.name "Your Name"
# git config --global user.email "you@example.com"
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בואו נראה מה כבר מוגדר אצלכם. התא הבא רק <strong>קורא</strong> את ההגדרות. אם אין הגדרה, הוא יאמר את זה בלי להיכשל:
</p>

```python
import subprocess

for שדה in ["user.name", "user.email"]:
    תוצאה = subprocess.run(["git", "config", "--global", "--get", שדה],
                           capture_output=True, text=True, encoding="utf-8")
    ערך = תוצאה.stdout.strip()
    print(שדה, "=", ערך if ערך else "(עדיין לא הוגדר)")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
הפקודה <code>git config --global --get user.name</code> מחזירה את הערך השמור. כשאין ערך היא לא מדפיסה כלום, ולכן הוספנו הודעה ידידותית.<br>
(בקוד השתמשנו ב-<code>subprocess</code>, כלי של פייתון להרצת פקודות. אפשר גם פשוט <code>!git config --global --get user.name</code>.)<br>
<strong>במחברת הזו לא נצטרך זהות גלובלית.</strong> בתיקיית הניסויים נגדיר זהות מקומית, שנשמרת רק שם.
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">תיקיית ניסויים (Sandbox)</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
כדי להתנסות בלי לגעת בקבצים אמיתיים, המחברת תיצור תיקייה חדשה בשם <code>git_playground</code> לצד המחברת, ותעבוד בתוכה.<br>
התא הבא עושה כמה דברים:
</p>

<ol style="text-align: right; direction: rtl; float: right; clear: both;">
    <li>אם התיקייה כבר קיימת מהרצה קודמת - מוחק אותה. כך אפשר להריץ את המחברת שוב ושוב מההתחלה.</li>
    <li>יוצר תיקייה חדשה וריקה.</li>
    <li><strong>נכנס אליה</strong> באמצעות <code>os.chdir</code>.</li>
</ol>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>למה צריך להיכנס לתיקייה?</strong> פקודות git פועלות על התיקייה שבה הן רצות (התיקייה הנוכחית).<br>
<code>os.chdir("שם")</code> של פייתון משנה את התיקייה הנוכחית (כמו <code>cd</code> בחלון פקודות; ב-Jupyter קיים גם <code>%cd</code> שעושה אותו דבר).<br>
שימו לב: <code>!cd שם</code> <strong>לא</strong> עובד, כי כל שורה עם <code>!</code> רצה בחלון פקודות חדש משלה ושוכחת את <code>cd</code> מיד.<br>
אחרי התא הזה, <strong>כל פקודת <code>!git</code> במחברת רצה "בתוך" תיקיית הניסויים</strong>.<br>
את הקבצים ניצור באמצעות פייתון (ולא <code>echo</code>), כי כך זה עובד זהה ב-Windows, Mac ו-Linux.
</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
הערה טכנית: בתוך תיקיית <code>.git</code> יש קבצים שמסומנים "לקריאה בלבד", וב-Windows מחיקה רגילה שלהם נכשלת. לכן פונקציית המחיקה שלנו מסירה את הסימון לפני המחיקה.
</p>

```python
import os
import shutil
import stat
import subprocess
import sys

# שמות התיקיות שהמחברת יוצרת וגם מוחקת. אל תשמרו שם עבודה אמיתית!
שמות_ארגז = ["git_playground", "git_playground_remote", "git_playground_clone"]

# אם הרצנו את התא הזה שוב כשאנחנו בתוך אחת התיקיות - קודם יוצאים החוצה
if os.path.basename(os.getcwd()) in שמות_ארגז + ["git_exercise"]:
    os.chdir("..")
תיקיית_המחברת = os.getcwd()


def טיפול_בקובץ_מוגן(פעולה, נתיב, _):
    """ב-Windows קבצים בתוך .git הם 'לקריאה בלבד': מסירים את הסימון ומנסים שוב."""
    os.chmod(נתיב, stat.S_IWRITE)
    פעולה(נתיב)


def מחק_תיקייה(נתיב):
    """מוחקת תיקייה (אם קיימת), כולל תיקיות .git."""
    if not os.path.exists(נתיב):
        return
    if sys.version_info >= (3, 12):
        shutil.rmtree(נתיב, onexc=טיפול_בקובץ_מוגן)
    else:
        shutil.rmtree(נתיב, onerror=טיפול_בקובץ_מוגן)


def כתוב_קובץ(נתיב, תוכן):
    """יוצרת (או דורסת) קובץ טקסט בקידוד utf-8."""
    with open(נתיב, "w", encoding="utf-8", newline="\n") as f:
        f.write(תוכן)


def הצג_קובץ(נתיב):
    """מדפיסה את תוכן הקובץ."""
    with open(נתיב, encoding="utf-8") as f:
        print(f.read())


def פלט_git(*ארגומנטים):
    """מריצה פקודת git ומחזירה את הפלט שלה כטקסט (לשימוש בתוך פייתון)."""
    תוצאה = subprocess.run(["git", *ארגומנטים], capture_output=True, text=True, encoding="utf-8")
    return תוצאה.stdout.strip()


if shutil.which("git") is None:
    print("git לא נמצא במחשב, אז לא ניצור תיקיית ניסויים. אפשר להמשיך לקרוא בלי להריץ.")
else:
    for שם in שמות_ארגז:
        מחק_תיקייה(שם)
    os.mkdir("git_playground")
    os.chdir("git_playground")
    print("התיקייה הנוכחית היא עכשיו:")
    print(os.getcwd())
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
המחברת מחקה תיקיות ישנות (אם היו), יצרה את <code>git_playground</code> ונכנסה אליה.<br>
הדפסנו את הנתיב המלא כדי שתראו איפה אתם. הוא נגמר ב-<code>git_playground</code>.<br>
בתא הזה הגדרנו גם כמה פונקציות עזר קטנות שנשתמש בהן: <code>כתוב_קובץ</code> (יוצרת קובץ), <code>הצג_קובץ</code> (מדפיסה קובץ) ו-<code>פלט_git</code> (מריצה git ומחזירה את התוצאה לפייתון).<br>
<strong>הרצת התא הזה שוב בכל שלב מחזירה אתכם להתחלה נקייה.</strong>
</p>

### <p style="text-align: right; direction: rtl; float: right; clear: both;">יצירת מאגר: git init</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
כדי שתיקייה תהפוך ל"פרויקט git", מריצים בתוכה <code>git init</code> (מלשון initialize - אתחול).<br>
הפקודה יוצרת בתוך התיקייה תת-תיקייה נסתרת בשם <code>.git</code>. <strong>שם git שומר את כל ההיסטוריה.</strong><br>
(אם תמחקו את <code>.git</code>, תישאר תיקייה רגילה, בלי היסטוריה.)
</p>

```python
!git init
```

הפקודה השנייה אמרה ל-HEAD להצביע על ענף בשם <code>main</code>, והשלישית הדפיסה את שם הענף הנוכחי: <code>main</code>.<br>

השורה האחרונה מבטיחה שהדוגמאות יעבדו גם אם במחשב שלכם מוגדרת חתימה אוטומטית על commits.

## <p style="text-align: right; direction: rtl; float: right; clear: both;">המחזור הבסיסי: status, add, commit</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
את רוב העבודה עם git עושים בלולאה קבועה: <strong>עורכים קבצים ← בודקים מה השתנה ← בוחרים מה לשמור (add) ← שומרים (commit)</strong>.<br>
נבנה פרויקט קטן: סקריפט שמחשב דוח מכירות.
</p>

### <p style="text-align: right; direction: rtl; float: right; clear: both;">git status - איפה אני עומד?</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
הפקודה הכי שימושית ב-git היא <code>git status</code>. היא אומרת מה המצב: באיזה ענף אתם, מה השתנה, ומה מוכן לשמירה.<br>
כשלא בטוחים - מריצים <code>git status</code>. אין בה שום סכנה, היא רק קוראת ולא משנה כלום.
</p>

```python
!git status
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
<code>On branch main</code> - אנחנו בענף <code>main</code>.<br>
<code>No commits yet</code> - עדיין לא שמרנו אף גרסה.<br>
<code>nothing to commit</code> - אין מה לשמור, כי התיקייה ריקה.
</p>

### <p style="text-align: right; direction: rtl; float: right; clear: both;">יצירת קובץ ובדיקה מחדש</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
ניצור את הגרסה הראשונה של הסקריפט, באמצעות הפונקציה <code>כתוב_קובץ</code> שהגדרנו (שפותחת קובץ וכותבת לתוכו, בדיוק כמו <code>open(..., "w")</code>), ונריץ אותו לראות שהוא עובד.
</p>

```python
כתוב_קובץ("sales_report.py", """# sales_report.py - daily sales report
sales = [120, 340, 95, 410]

total = sum(sales)
print("Total sales:", total)
""")

# מריצים את הסקריפט כמו שמריצים קובץ פייתון
%run sales_report.py
```

```python
!git status
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
הקובץ החדש מופיע תחת <code>Untracked files</code>, כלומר "קבצים ש-git לא עוקב אחריהם".<br>
git רואה שהקובץ קיים בתיקייה, אבל עדיין לא קיבל הוראה לשמור אותו.<br>
שימו לב שהוא גם מציע לנו מה לעשות: <code>use "git add &lt;file&gt;..." to include in what will be committed</code>. הודעות הסטטוס של git הן בעצם מדריך קצר.
</p>

### <p style="text-align: right; direction: rtl; float: right; clear: both;">git add - מכניסים לאזור ההכנה</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<code>git add שם_קובץ</code> מעביר את הקובץ מתיקיית העבודה לאזור ההכנה. עדיין לא שמרנו כלום - רק סימנו מה ייכנס ל-commit הבא.
</p>

```python
!git add sales_report.py
!git status
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
עכשיו הקובץ מופיע תחת <code>Changes to be committed</code> עם התווית <code>new file</code>.<br>
זה אומר שהוא באזור ההכנה, ומוכן להישמר. הפקודה <code>git add</code> עצמה לא מדפיסה כלום כשהיא מצליחה - שתיקה היא סימן טוב.
</p>

### <p style="text-align: right; direction: rtl; float: right; clear: both;">git commit - שומרים גרסה</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<code>git commit -m "הודעה"</code> לוקח את כל מה שנמצא באזור ההכנה ושומר אותו כ-commit חדש בהיסטוריה. הדגל <code>-m</code> (message) מוסיף את ההודעה.<br>
<strong>חשוב:</strong> תמיד נכתוב <code>-m</code> עם הודעה בגרשיים כפולות. בלי <code>-m</code>, git פותח עורך טקסט, ובמחברת (ובמיוחד ב-Windows) זה עלול לתקוע הכול.
</p>

```python
!git commit -m "Add first version of sales report"
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
השורה הראשונה בפלט: <code>[main (root-commit) 1a2b3c4] Add first version...</code>.<br>
<code>main</code> הוא הענף, <code>root-commit</code> אומר שזה ה-commit הראשון (בלי הורה), והקוד שאחריו (7 תווים אצלכם יהיו שונים משלי) הוא המזהה של ה-commit.<br>
בהמשך git מסכם: קובץ אחד השתנה, וכמה שורות נוספו.
</p>

```python
!git status
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
<code>nothing to commit, working tree clean</code> - "עץ העבודה נקי": כל מה שבתיקייה זהה למה ששמור ב-commit האחרון.<br>
זה המצב שאנחנו רוצים להגיע אליו אחרי כל שמירה.
</p>

### <p style="text-align: right; direction: rtl; float: right; clear: both;">git log - רואים את ההיסטוריה</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<code>git log</code> מציג את ה-commits, <strong>מהחדש לישן</strong>. לכל commit יש מזהה מלא, מחבר, תאריך והודעה.<br>
עם <code>--oneline</code> מקבלים שורה אחת לכל commit (מזהה מקוצר + הודעה). זה נוח כשיש הרבה.
</p>

```python
!git log
```

```python
!git log --oneline
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
בפקודה הראשונה רואים את כל הפרטים: <code>commit</code> (המזהה המלא), <code>Author</code> (Dana Cohen, כפי שהגדרנו), <code>Date</code> וההודעה.<br>
בפקודה השנייה מקבלים אותו commit בשורה אחת. כרגע יש רק אחד, אבל בקרוב יהיו יותר.<br>
ייתכן שליד המזהה יופיע גם <code>(HEAD -&gt; main)</code>: זה סימון של git שאומר "אתם נמצאים כאן, בענף main". תלוי בסביבה, הוא מופיע אצל חלק מכם ואצל חלק לא, ושני המצבים תקינים.
</p>

### <p style="text-align: right; direction: rtl; float: right; clear: both;">עורכים קובץ ורואים מה השתנה: git diff</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
נוסיף לסקריפט חישוב של ממוצע. נשנה את הקובץ ואז נשאל את git מה השתנה.
</p>

```python
כתוב_קובץ("sales_report.py", """# sales_report.py - daily sales report
sales = [120, 340, 95, 410]

total = sum(sales)
average = total / len(sales)
print("Total sales:", total)
print("Average sale:", average)
""")

%run sales_report.py
```

```python
!git status
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
הפעם הקובץ מופיע תחת <code>Changes not staged for commit</code> עם התווית <code>modified</code>.<br>
זה אומר: git מכיר את הקובץ, הוא השתנה מאז ה-commit האחרון, אבל השינוי עדיין <strong>לא</strong> באזור ההכנה.
</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
כדי לראות <strong>מה בדיוק</strong> השתנה, משתמשים ב-<code>git diff</code> (מלשון difference - הבדל):
</p>

```python
!git diff
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
קורא את הפלט ככה:<br>
שורות שמתחילות ב-<code>+</code> (ירוק במסוף צבעוני) - שורות שנוספו. שורות עם <code>-</code> - שורות שנמחקו. שורות בלי סימן - הקשר, כדי להבין איפה זה.<br>
כאן נוספו שורה של <code>average</code> ושורת הדפסה. <code>@@ ... @@</code> אומר באיזו שורה בקובץ זה קרה.<br>
<code>git diff</code> בלי שום דבר נוסף משווה בין <strong>תיקיית העבודה לאזור ההכנה</strong>: מראה שינויים שעוד לא הוכנסו.
</p>

### <p style="text-align: right; direction: rtl; float: right; clear: both;">git diff --staged - מה עומד להישמר?</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
נכניס את השינוי לאזור ההכנה. עכשיו <code>git diff</code> ריק (אין שינויים שלא הוכנסו), ואילו <code>git diff --staged</code> מראה מה נמצא באזור ההכנה, כלומר <strong>מה ייכנס ל-commit הבא</strong>.<br>
זו בדיקה טובה לפני כל commit.
</p>

```python
!git add sales_report.py
print("--- git diff (שינויים שלא הוכנסו): ---")
!git diff
print("--- git diff --staged (שינויים שהוכנסו): ---")
!git diff --staged
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
הראשון הוציא פלט ריק, כי כבר העברנו הכול לאזור ההכנה.<br>
השני הציג את אותם שינויים כמו קודם - הם מחכים ל-commit.<br>
חשבו על זה כך: <code>git diff</code> = "מה עשיתי ולא הכנתי", <code>git diff --staged</code> = "מה הכנתי ולא שמרתי".
</p>

### <p style="text-align: right; direction: rtl; float: right; clear: both;">3. מחזירים גרסה ישנה של קובץ: git restore --source</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">

```python
קוד_ראשון = פלט_git("rev-list", "--max-parents=0", "HEAD")[:7]
print("המזהה של ה-commit הראשון:", קוד_ראשון)
```

הנקודה החשובה: <code>--source</code> שינה את <strong>תיקיית העבודה</strong> בלבד. ההיסטוריה לא השתנתה, וה-commits החדשים ממשיכים להיות שמורים.<br>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">ענפים (Branches)</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
נניח שאתם רוצים לנסות רעיון: להוסיף לדוח את המכירה הגדולה ביותר. אתם לא בטוחים שזה יצליח, ובינתיים רוצים שהגרסה שעובדת תישאר שלמה.<br>
בשביל זה יש <strong>ענפים</strong> (branches): קו עבודה נפרד, שאפשר לנסות בו דברים, ואם הוא טוב - לחבר אותו חזרה לקו הראשי.

(git מסרב לעבור ענף אם יש שינויים שלא נשמרו שיידרסו. אז עושים commit לפני המעבר.)

### <p style="text-align: right; direction: rtl; float: right; clear: both;">מחברים את הענף חזרה: git merge</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
הרעיון עבד, והענף מוכן. כדי להביא את השינויים שלו ל-<code>main</code>, <strong>עוברים ל-main</strong> ומבקשים <code>git merge שם_הענף</code> (merge = מיזוג).<br>
זה המקרה הפשוט: ב-<code>main</code> לא נעשה כלום מאז שפרשנו, ולכן git רק "מזיז את התווית קדימה". זה נקרא <strong>fast-forward</strong>.
</p>

```python
!git switch main
!git merge feature-top-sale
!git log --oneline --graph --all --decorate
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
ההודעה <code>Fast-forward</code> אומרת בדיוק את זה: <code>main</code> התקדם ישר אל ה-commit של הענף, בלי commit חדש.<br>
בגרף שני הענפים מצביעים עכשיו על אותו commit.
</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
הענף כבר מיזגנו ואין בו צורך. נמחק אותו (המחיקה מסירה רק את ה<strong>תווית</strong>; ה-commits נשארים בהיסטוריה). האות הקטנה <code>-d</code> בטוחה: git מסרב למחוק ענף שעוד לא מוזג.
</p>

```python
!git branch -d feature-top-sale
!git branch
```

מה קורה אם גם <code>main</code> התקדם בזמן שעבדנו בענף? אז אי אפשר פשוט להזיז תווית. git <strong>מחבר את שני הקווים</strong> ויוצר commit מיוחד של מיזוג (merge commit) עם שני הורים.<br>
נדגים: נפתח ענף לחישוב מע"מ, ובינתיים נוסיף ל-<code>main</code> קובץ README.

<p style="text-align: right; direction: rtl; float: right; clear: both;">
עכשיו ב-<code>main</code> יש commit שאין ב-<code>feature-vat</code>, ולהפך. נמזג. הוספנו <code>-m</code> כדי ש-git לא יפתח עורך להודעת המיזוג:

הפעם אין fast-forward: git כתב משהו כמו <code>Merge made by the 'ort' strategy</code> (בגרסאות ישנות: 'recursive'), ויצר commit חדש של מיזוג.<br>
הגרף מראה שני קווים שנפרדו ונפגשו מחדש. ל-merge commit יש שני הורים.<br>
git הצליח למזג לבד, כי השינויים נגעו בקבצים שונים (README מול הסקריפט).
</p>

```python
!git branch -d feature-vat
```

### <p style="text-align: right; direction: rtl; float: right; clear: both;">התנגשות (Conflict) - זה נורמלי</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
מה אם שני הענפים שינו <strong>את אותה שורה</strong> בצורות שונות? git לא יכול לדעת מי צודק, ולכן הוא עוצר ומבקש שאתם תחליטו. זה נקרא <strong>התנגשות</strong> (merge conflict).<br>
<strong>התנגשות היא לא תקלה ולא טעות שלכם.</strong> היא חלק רגיל מעבודה עם git, במיוחד כשכמה אנשים עובדים על אותו קובץ. נעשה אחת בכוונה, כדי שתכירו אותה מקרוב.<br>
נשנה את הכותרת בקובץ README בשני ענפים, כל אחד בצורה אחרת:
</p>

<code>Automatic merge failed; fix conflicts and then commit the result.</code> - המיזוג האוטומטי נכשל; תקנו את ההתנגשות ואז עשו commit.<br>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">שיתוף עם אחרים: remote, clone, push, pull</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
עד עכשיו הכול קרה במחשב אחד. אבל כוחו האמיתי של git הוא בשיתוף. המושגים:
</p>

<ul style="text-align: right; direction: rtl; float: right; clear: both;">
    <li><strong>Remote</strong> (מאגר מרוחק): עותק נוסף של המאגר, שנמצא במקום אחר (בדרך כלל שרת כמו GitHub). בדרך כלל נותנים לו את הכינוי <code>origin</code>.</li>
    <li><strong>clone</strong>: יצירת עותק מלא של מאגר קיים (כולל כל ההיסטוריה) במחשב שלכם.</li>
    <li><strong>push</strong>: שליחת ה-commits שלכם אל ה-remote.</li>
    <li><strong>pull</strong>: הבאת commits חדשים מה-remote אל המחשב שלכם ומיזוג שלהם.</li>
</ul>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
כדי שנוכל להתנסות בלי אינטרנט ובלי חשבון GitHub, נשתמש בטריק: <strong>תיקייה רגילה בתוך המחשב שתשמש כ"שרת"</strong>. מבחינת git אין הבדל - כתובת של remote יכולה להיות גם נתיב בתיקיות.<br>
תיקייה כזו נוצרת עם <code>git init --bare</code>: מאגר "חשוף", בלי תיקיית עבודה, רק היסטוריה. זה בדיוק מה ששרתים כמו GitHub מחזיקים.
</p>

```python
!git init --bare ../git_playground_remote
# כמו קודם, קובעים שהענף הראשי של ה"שרת" הוא main (GitHub עושה את זה לבד)
!git --git-dir=../git_playground_remote symbolic-ref HEAD refs/heads/main
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
נוצרה תיקייה חדשה <code>git_playground_remote</code> לצד התיקייה שלנו (<code>..</code> אומר "התיקייה שמעל"). זה ה"שרת" שלנו, והוא ריק.<br>
הפרמטר <code>--git-dir</code> אמר ל-git להפעיל את הפקודה על המאגר הזה במקום על התיקייה הנוכחית.
</p>

### <p style="text-align: right; direction: rtl; float: right; clear: both;">מחברים את המאגר ל-remote: git remote add</p>

```python
!git remote add origin ../git_playground_remote
!git remote -v
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
<code>git remote add origin כתובת</code> אומר: "הכינוי <code>origin</code> מציין את המאגר שנמצא בכתובת הזו".<br>
<code>git remote -v</code> מציג את הרשימה: פעם עבור fetch (הבאה) ופעם עבור push (שליחה). הכתובת כאן היא נתיב, ובאמת ב-GitHub זו תהיה כתובת אינטרנט.
</p>

### <p style="text-align: right; direction: rtl; float: right; clear: both;">שולחים: git push</p>

```python
!git push -u origin main
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
<code>git push origin main</code> שולח את הענף <code>main</code> אל <code>origin</code>. ההודעות מראות כמה אובייקטים נשלחו, ואת השורה <code>[new branch] main -&gt; main</code>.<br>
הדגל <code>-u</code> (מלשון upstream) זוכר שהענף <code>main</code> שלנו מקושר ל-<code>origin/main</code>. מעכשיו מספיק לכתוב <code>git push</code> ו-<code>git pull</code> בלי פרטים נוספים.<br>
זו גם הפעולה שמגבה את העבודה שלכם מחוץ למחשב.<br>
(ייתכן שיופיעו גם שורות התקדמות כמו "Counting objects" ו-"Compressing objects" - הן רק מדווחות מה git עושה, ואין צורך להבין אותן.)
</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">

בכל מקרה: <strong>לעולם לא כותבים סיסמה או טוקן בתוך קובץ או בתוך הקוד, ולא מדביקים אותם במחברת.</strong> GitHub מבקש התחברות דרך הדפדפן, ו-git שומר את האישור בעצמו.

### <p style="text-align: right; direction: rtl; float: right; clear: both;">הקשר לקורס: מה קורה כשעושים git pull</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
הקורס שאתם משתמשים בו הוא בעצמו מאגר git. כשהמרצה מעדכן מחברת, הוא עושה commit ו-push, ואתם מריצים <code>git pull</code> כדי לקבל את העדכון.<br>

אם המחברת עודכנה גם אצל המרצה, ה-pull עלול להיכשל עם הודעה כמו:<br>
<code>error: Your local changes to the following files would be overwritten by merge</code>
</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
git מסרב לדרוס את העבודה שלכם, וזה טוב. <strong>מה עושים?</strong> הדרך הכי פשוטה, בלי להבין עוד פקודות:
</p>

<ol style="text-align: right; direction: rtl; float: right; clear: both;">

    <li>מחזירים את המחברת המקורית למצבה: <code>git restore content/day06/1_Pandas_Basics.ipynb</code> (שימו לב: זה זורק את השינויים בה, ולכן העתקתם קודם!).</li>
