## <p style="text-align: right; direction: rtl; float: right; clear: both;">סינון נתונים</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
אחת הפעולות הנפוצות ביותר בניתוח נתונים היא <strong>סינון</strong>.<br>
אנחנו רוצים למצוא רק את השורות שעומדות בתנאי מסוים.
</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>דוגמאות לשאלות שסינון עונה עליהן:</strong>
</p>

<ul style="text-align: right; direction: rtl; float: right; clear: both;">
    <li>מי כל העובדים שמשכורתם מעל 15,000?</li>
    <li>אילו לקוחות גרים בתל אביב?</li>
    <li>מהם כל המוצרים שנמכרו ביוני?</li>
</ul>

```python
import pandas as pd

# יצירת דוגמה
data = {
    'שם': ['דני', 'שרה', 'יוסי', 'מיכל', 'רועי'],
    'גיל': [25, 30, 28, 35, 22],
    'עיר': ['תל אביב', 'חיפה', 'ירושלים', 'תל אביב', 'חיפה'],
    'משכורת': [10000, 15000, 12000, 18000, 9000]
}

df = pd.DataFrame(data)
print("הטבלה המלאה:")
print(df)
```

### <p style="text-align: right; direction: rtl; float: right; clear: both;">סינון פשוט</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
הדרך הבסיסית לסנן היא ליצור "מסכה" של True/False:
</p>

```python
# יצירת מסכה - מערך של True/False
mask = df['גיל'] > 25
print("המסכה:")
print(mask)

print("\n" + "="*50 + "\n")

# שימוש במסכה לסינון
employees_over_25 = df[mask]
print("עובדים מעל גיל 25:")
print(employees_over_25)
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
כשכתבנו <code>df['גיל'] > 25</code>, Pandas בדק לכל שורה האם הגיל גדול מ-25.<br>
קיבלנו מערך של True/False - True בשורות שעומדות בתנאי.<br>
כשהעברנו את המסכה הזו ב-<code>df[mask]</code>, קיבלנו רק את השורות עם True!
</p>

### <p style="text-align: right; direction: rtl; float: right; clear: both;">דרך קצרה</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בפועל, לא צריך משתנה נפרד למסכה. אפשר לכתוב הכל בשורה אחת:
</p>

```python
# סינון ישיר - הדרך הנפוצה ביותר
high_salary = df[df['משכורת'] > 12000]
print("עובדים עם משכורת מעל 12,000:")
print(high_salary)
```

### <p style="text-align: right; direction: rtl; float: right; clear: both;">תנאים מרובים - AND</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
לפעמים נרצה שכמה תנאים יתקיימו <strong>יחד</strong>.<br>
נשתמש ב-<code>&</code> (AND) ונשים כל תנאי בסוגריים:
</p>

```python
# סינון עם 2 תנאים (AND)
# חייבים סוגריים סביב כל תנאי!
result = df[(df['גיל'] > 25) & (df['עיר'] == 'תל אביב')]
print("עובדים מעל 25 שגרים בתל אביב:")
print(result)
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>שימו לב!</strong> חובה להשתמש ב-<code>&</code> ולא ב-<code>and</code> של Python הרגיל.<br>
וחובה לשים סוגריים סביב כל תנאי! אחרת תקבלו שגיאה.
</p>

### <p style="text-align: right; direction: rtl; float: right; clear: both;">תנאים מרובים - OR</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
אם נרצה ש<strong>לפחות אחד</strong> מהתנאים יתקיים, נשתמש ב-<code>|</code> (OR):
</p>

```python
# סינון עם OR - מספיק שתנאי אחד מתקיים
result = df[(df['עיר'] == 'חיפה') | (df['עיר'] == 'תל אביב')]
print("עובדים מחיפה או תל אביב:")
print(result)
```

### <p style="text-align: right; direction: rtl; float: right; clear: both;">isin - בדיקת השתייכות לרשימה</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
כשיש לנו הרבה אפשרויות, <code>isin()</code> יותר נוח מ-OR רבים:
</p>

```python
# דרך יפה יותר עם isin
cities = ['חיפה', 'תל אביב']
result = df[df['עיר'].isin(cities)]
print("עובדים מהערים הנבחרות:")
print(result)
```

## <p style="text-align: right; direction: rtl; float: right; clear: both;">מיון נתונים</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
לפעמים נרצה למיין את הטבלה לפי עמודה מסוימת.
</p>

```python
# מיון לפי גיל (עולה)
print("ממוין לפי גיל:")
print(df.sort_values('גיל'))

print("\n" + "="*50 + "\n")

# מיון לפי משכורת (יורד)
print("ממוין לפי משכורת (גבוה לנמוך):")
print(df.sort_values('משכורת', ascending=False))
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>פרמטרים חשובים:</strong><br>
<code>ascending=True</code> (ברירת מחדל) - מיון עולה (קטן לגדול, A-Z).<br>
<code>ascending=False</code> - מיון יורד (גדול לקטן, Z-A).
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">עמודות מחושבות</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
אפשר ליצור עמודה חדשה מתוך עמודות קיימות, בדיוק כמו משתנה חדש.<br>
Pandas מבצע את החישוב על כל השורות בבת אחת - בלי לולאה!
</p>

```python
# עמודה חדשה: משכורת שנתית
df['משכורת_שנתית'] = df['משכורת'] * 12

# עמודה של True/False: האם המשכורת גבוהה?
df['משכורת_גבוהה'] = df['משכורת'] >= 15000

print(df)
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
כתבנו <code>df['שם_עמודה_חדש'] = ...</code> והעמודה נוספה לטבלה.<br>
החישוב <code>df['משכורת'] * 12</code> רץ על כל השורות ביחד.<br>
שימו לב: ליצירת עמודה חדשה חייבים סוגריים מרובעים, נקודה לא תעבוד.
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">ספירת ערכים - value_counts</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
שאלה מאוד נפוצה: "כמה פעמים מופיע כל ערך בעמודה?" - כמה עובדים בכל עיר, כמה הזמנות מכל מוצר.<br>
<code>value_counts()</code> עונה על זה בשורה אחת, וממיין מהנפוץ לנדיר:
</p>

```python
print("כמה עובדים בכל עיר:")
print(df['עיר'].value_counts())
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
קיבלנו טבלה קטנה: לכל עיר, כמה שורות יש בה. זה קיצור של <code>groupby('עיר').size()</code> עם מיון.
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">קיבוץ - groupby</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<code>groupby</code> היא אחת הפונקציות החזקות ביותר ב-Pandas!<br>
היא מאפשרת לקבץ שורות לפי ערך משותף ולבצע חישובים על כל קבוצה.
</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>למשל:</strong>
</p>

<ul style="text-align: right; direction: rtl; float: right; clear: both;">
    <li>מה הממוצע של משכורות לפי עיר?</li>
    <li>כמה עובדים יש בכל מחלקה?</li>
    <li>מה סכום המכירות לפי חודש?</li>
</ul>

```python
# ממוצע משכורת לפי עיר
print("ממוצע משכורת לפי עיר:")
print(df.groupby('עיר')['משכורת'].mean())

print("\n" + "="*50 + "\n")

# ספירה - כמה עובדים בכל עיר
print("מספר עובדים לפי עיר:")
print(df.groupby('עיר').size())
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>איך זה עובד?</strong><br>
<code>groupby('עיר')</code> - מקבץ את כל השורות שיש להן אותה עיר.<br>
<code>['משכורת']</code> - בוחר את עמודת המשכורת.<br>
<code>.mean()</code> - מחשב ממוצע לכל קבוצה.
</p>

### <p style="text-align: right; direction: rtl; float: right; clear: both;">פונקציות אגרגציה נפוצות</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
אחרי <code>groupby</code> אפשר להשתמש בפונקציות רבות:
</p>

```python
# מינימום, מקסימום, סכום
print("משכורת מקסימלית לפי עיר:")
print(df.groupby('עיר')['משכורת'].max())

print("\n" + "="*50 + "\n")

print("סכום משכורות לפי עיר:")
print(df.groupby('עיר')['משכורת'].sum())
```

### <p style="text-align: right; direction: rtl; float: right; clear: both;">agg - כמה חישובים ביחד</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
אם רוצים כמה סטטיסטיקות בבת אחת, משתמשים ב-<code>agg()</code>:
</p>

```python
# כמה חישובים ביחד
print(df.groupby('עיר')['משכורת'].agg(['mean', 'min', 'max', 'count']))
```

## <p style="text-align: right; direction: rtl; float: right; clear: both;">גרף ראשון - df.plot</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
מספרים בטבלה קשה לקרוא, וגרף עושה את זה מיד ברור.<br>
Pandas משתמש בספרייה <code>matplotlib</code> ומציג גרף בשורה אחת: <code>.plot()</code>.<br>
נטען את קובץ המכירות מתיקיית <code>resources</code>, נחשב סכום מכירות לכל חודש ונציג גרף עמודות:
</p>

```python
import matplotlib.pyplot as plt

sales = pd.read_csv('../../resources/sales.csv')

# עמודה מחושבת: החודש (למשל 2024-01) מתוך התאריך
sales['month'] = pd.to_datetime(sales['date']).dt.strftime('%Y-%m')

# סכום המכירות לכל חודש
by_month = sales.groupby('month')['total'].sum()
print(by_month)

# גרף עמודות
by_month.plot(kind='bar', title='Total sales per month')
plt.ylabel('Total (NIS)')
plt.show()
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
<code>groupby('month')['total'].sum()</code> נתן לנו מספר אחד לכל חודש, ו-<code>.plot(kind='bar')</code> צייר אותו כעמודות.<br>
<code>plt.show()</code> מציג את הגרף. <code>kind</code> יכול להיות גם <code>'line'</code> (קו) או <code>'pie'</code> (עוגה).
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">מיזוג טבלאות - merge</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
הרבה פעמים הנתונים שלנו מפוזרים בכמה טבלאות.<br>
<code>merge</code> מאפשר לחבר טבלאות לפי עמודה משותפת - כמו JOIN ב-SQL.
</p>

```python
# טבלה 1 - פרטי עובדים
employees = pd.DataFrame({
    'id': [1, 2, 3],
    'שם': ['דני', 'שרה', 'יוסי'],
    'מחלקה_id': [10, 20, 10]
})

# טבלה 2 - פרטי מחלקות
departments = pd.DataFrame({
    'id': [10, 20, 30],
    'שם_מחלקה': ['פיתוח', 'שיווק', 'מכירות']
})

print("טבלת עובדים:")
print(employees)
print("\n" + "="*50 + "\n")
print("טבלת מחלקות:")
print(departments)
```

```python
# מיזוג - חיבור העובדים למחלקות שלהם
result = pd.merge(
    employees,
    departments,
    left_on='מחלקה_id',  # העמודה בטבלה השמאלית
    right_on='id',        # העמודה בטבלה הימנית
    how='left'            # סוג המיזוג
)

print("\nאחרי מיזוג:")
print(result)
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>סוגי מיזוג (how):</strong>
</p>

<ul style="text-align: right; direction: rtl; float: right; clear: both;">
    <li><strong>left</strong> - שומר את כל השורות מהטבלה השמאלית</li>
    <li><strong>right</strong> - שומר את כל השורות מהטבלה הימנית</li>
    <li><strong>inner</strong> (ברירת מחדל) - רק שורות שיש להן התאמה בשתי הטבלאות</li>
    <li><strong>outer</strong> - כל השורות משתי הטבלאות</li>
</ul>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">טיפול בערכים חסרים</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
בעולם האמיתי, לפעמים יש ערכים חסרים בנתונים.<br>
Pandas מסמן אותם כ-<code>NaN</code> (Not a Number).
</p>

```python
import numpy as np

# יצירת DataFrame עם ערכים חסרים
df_missing = pd.DataFrame({
    'שם': ['דני', 'שרה', 'יוסי'],
    'גיל': [25, np.nan, 28],
    'עיר': ['תל אביב', 'חיפה', None]
})

print("נתונים עם ערכים חסרים:")
print(df_missing)

print("\n" + "="*50 + "\n")

# בדיקה איפה יש ערכים חסרים
print("האם יש ערכים חסרים?")
print(df_missing.isnull())
```

```python
# מילוי ערכים חסרים
filled = df_missing.fillna({
    'גיל': df_missing['גיל'].mean(),  # ממוצע
    'עיר': 'לא ידוע'
})

print("אחרי מילוי:")
print(filled)

print("\n" + "="*50 + "\n")

# מחיקת שורות עם ערכים חסרים
cleaned = df_missing.dropna()
print("אחרי מחיקת שורות חסרות:")
print(cleaned)
```
