```python
import os
import time
import threading
import requests

os.makedirs('my_images', exist_ok=True)


def download_image(url, filename):
    """מורידה תמונה אחת ושומרת אותה. מחזירה True אם הצליח"""
    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status()
        with open(filename, 'wb') as f:
            f.write(response.content)
        return True
    except Exception as e:
        print(f"שגיאה ב-{filename}: {e}")
        return False


# רשימת (כתובת, שם קובץ)
urls = [(f"https://picsum.photos/300/200?random={n}", f"my_images/img{n}.jpg") for n in range(1, 11)]

# הקוד שלכם כאן: הורדה רגילה, הורדה עם threads, והשוואת זמנים
```
