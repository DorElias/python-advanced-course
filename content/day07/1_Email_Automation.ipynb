## <p style="text-align: right; direction: rtl; float: right; clear: both;">שליחת מייל פשוט</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
נתחיל בדוגמה הכי בסיסית - שליחת מייל טקסט פשוט.
</p>

<div style="text-align: right; direction: rtl; clear: both; padding: 10px; background-color: #fff3cd; border-right: 4px solid #ffc107;">
    <strong>⚠️ חשוב!</strong><br>
    הקוד למטה הוא דוגמה. שליחה אמיתית דורשת App Password והפעלת אימות דו-שלבי, ולכן <strong>לא נשלח מייל אמיתי בכיתה</strong>.<br>את שורות השליחה השארנו בהערה (<code>#</code>) - הן מיועדות להרצה <strong>רק בבית</strong>, בחשבון שלכם.<br>בכיתה נבנה את המייל, נציג אותו ונשמור אותו לקובץ <code>.eml</code>.<br>שמרו סיסמאות בקובץ נפרד או במשתני סביבה - לעולם לא בקוד שמשתפים!

```python
import smtplib
from email.mime.text import MIMEText
from email.mime.multipart import MIMEMultipart

# פרטי החיבור (דוגמה - לא אמיתיים!)
SMTP_SERVER = 'smtp.gmail.com'
SMTP_PORT = 587
EMAIL = 'your-email@gmail.com'
PASSWORD = 'your-app-password'  # אל תשתמשו בסיסמה רגילה!

# יצירת המייל
msg = MIMEMultipart()
msg['From'] = EMAIL
msg['To'] = 'recipient@example.com'
msg['Subject'] = 'מייל ראשון מ-Python!'

# תוכן המייל
תוכן = '''
שלום,

זהו מייל אוטומטי שנשלח מסקריפט Python.
מגניב, נכון?

בברכה,
הבוט שלי
'''

msg.attach(MIMEText(תוכן, 'plain', 'utf-8'))

# שליחה אמיתית - רק בבית! (בכיתה נשאיר בהערה)
# try:
#     server = smtplib.SMTP(SMTP_SERVER, SMTP_PORT)
#     server.starttls()  # הצפנה
#     server.login(EMAIL, PASSWORD)
#     server.send_message(msg)
#     server.quit()
#     print("המייל נשלח בהצלחה!")
# except Exception as e:
#     print(f"שגיאה: {e}")

print("המייל נבנה בהצלחה (עדיין לא נשלח - שורות השליחה בהערה)")
```

### <p style="text-align: right; direction: rtl; float: right; clear: both;">מה יש בתוך המייל? - as_string()</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
המייל שבנינו הוא אובייקט בזיכרון. כדי לראות מה <strong>באמת</strong> יישלח לשרת, מדפיסים אותו עם <code>as_string()</code>.<br>
זה בטוח להרצה בכיתה - לא נשלח כלום:
</p>

```python
# מה שהשרת היה מקבל
print(msg.as_string())
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
ראינו את "הצד הטכני" של מייל: שורות header (<code>From</code>, <code>To</code>, <code>Subject</code>) ואחריהן התוכן.<br>
הנושא והתוכן בעברית מקודדים אוטומטית (<code>utf-8</code> ו-<code>base64</code>), ולכן הם נראים כמו אותיות מוזרות - זה תקין לגמרי,
וכל תוכנת מייל יודעת לפענח את זה. נפענח בעצמנו:
</p>

```python
# פענוח הטקסט מתוך החלק הראשון של המייל
חלק_ראשון = msg.get_payload()[0]
print(חלק_ראשון.get_payload(decode=True).decode('utf-8'))
```

### <p style="text-align: right; direction: rtl; float: right; clear: both;">שמירת מייל לקובץ eml</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
מכיוון שלא שולחים מיילים אמיתיים בכיתה, נשמור את המייל לקובץ עם סיומת <code>.eml</code>.<br>
זה פורמט סטנדרטי: לחיצה כפולה על הקובץ פותחת אותו ב-Outlook, Thunderbird או בתוכנת המייל של Windows/Mac -
בדיוק כמו מייל שהתקבל. מצב כתיבה <code>'wb'</code> כותב בייטים, ו-<code>as_bytes()</code> נותן את המייל בבייטים:
</p>

```python
# שמירת המייל לקובץ - עובד גם בלי אינטרנט
with open('first_mail.eml', 'wb') as f:
    f.write(msg.as_bytes())

print("נשמר הקובץ first_mail.eml - אפשר לפתוח אותו בתוכנת המייל")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
<code>open(..., 'wb')</code> פותח (או דורס) קובץ לכתיבה בייטים, ולכן אפשר להריץ שוב ושוב.<br>
כך נבדוק מיילים בבטחה, ורק כשהכל נראה טוב נחליף את השמירה בשליחה אמיתית.

## <p style="text-align: right; direction: rtl; float: right; clear: both;">שליחת מייל HTML מעוצב</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
מיילים יכולים להיות יותר מסתם טקסט!<br>
אפשר להוסיף עיצוב, צבעים, קישורים ועוד.
</p>

```python
from email.mime.text import MIMEText
from IPython.display import HTML, display
from email.mime.multipart import MIMEMultipart

# יצירת מייל
msg = MIMEMultipart('alternative')
msg['Subject'] = 'דוח שבועי - מערכת המכירות'
msg['From'] = 'system@company.com'
msg['To'] = 'manager@company.com'

# גרסת טקסט פשוט (לתמיכה במיילים שלא תומכים ב-HTML)
טקסט = '''
דוח שבועי
===========
סך מכירות: 45,000 ₪
מספר הזמנות: 23
'''

# גרסת HTML מעוצבת
html = '''
<html>
  <body dir="rtl">
    <h2 style="color: #2c3e50;">📊 דוח שבועי - מערכת המכירות</h2>
    <div style="background-color: #ecf0f1; padding: 20px; border-radius: 5px;">
      <p><strong>תקופה:</strong> 01/06/2024 - 07/06/2024</p>
      <table style="width: 100%; border-collapse: collapse;">
        <tr style="background-color: #3498db; color: white;">
          <th style="padding: 10px; text-align: right;">מדד</th>
          <th style="padding: 10px; text-align: right;">ערך</th>
        </tr>
        <tr>
          <td style="padding: 10px; border-bottom: 1px solid #bdc3c7;">סך מכירות</td>
          <td style="padding: 10px; border-bottom: 1px solid #bdc3c7; font-weight: bold; color: #27ae60;">45,000 ₪</td>
        </tr>
        <tr>
          <td style="padding: 10px; border-bottom: 1px solid #bdc3c7;">מספר הזמנות</td>
          <td style="padding: 10px; border-bottom: 1px solid #bdc3c7;">23</td>
        </tr>
        <tr>
          <td style="padding: 10px;">ממוצע להזמנה</td>
          <td style="padding: 10px;">1,956 ₪</td>
        </tr>
      </table>
    </div>
    <p style="color: #7f8c8d; font-size: 12px; margin-top: 20px;">
      מייל אוטומטי ממערכת הניהול | לא להשיב למייל זה
    </p>
  </body>
</html>
'''

# הוספת שתי הגרסאות
חלק_טקסט = MIMEText(טקסט, 'plain', 'utf-8')
חלק_html = MIMEText(html, 'html', 'utf-8')

msg.attach(חלק_טקסט)
msg.attach(חלק_html)

# שומרים את ה-HTML גם לקובץ - אפשר לפתוח אותו בדפדפן
with open('weekly_report.html', 'w', encoding='utf-8') as f:
    f.write(html)

print("מייל HTML מוכן! כך הוא יוצג אצל מי שיקבל אותו:")
display(HTML(html))
```

בנוסף שמרנו את הקובץ <code>weekly_report.html</code> - פתחו אותו בדפדפן כדי לראות אותו במסך מלא.

## <p style="text-align: right; direction: rtl; float: right; clear: both;">שליחת מייל עם קובץ מצורף</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
אחת הפונקציות השימושיות ביותר - צירוף קבצים!
</p>

```python
from email.mime.base import MIMEBase
from email import encoders
import os
from openpyxl import Workbook

# קודם יוצרים קובץ Excel קטן לדוגמה (כמו ביום 3), כדי שיהיה מה לצרף
wb = Workbook()
ws = wb.active
ws.title = 'יוני'
ws.append(['מוצר', 'כמות', 'מחיר'])
ws.append(['עכבר', 15, 50])
ws.append(['מקלדת', 10, 120])
wb.save('דוח_יוני.xlsx')


def בנה_מייל_עם_קובץ(נמען, נושא, תוכן, נתיב_קובץ):
    """
    בונה מייל עם קובץ מצורף (לא שולח אותו!)

    Args:
        נמען: כתובת המייל של הנמען
        נושא: נושא המייל
        תוכן: תוכן המייל
        נתיב_קובץ: נתיב לקובץ שרוצים לצרף
    """
    msg = MIMEMultipart()
    msg['From'] = 'sender@example.com'
    msg['To'] = נמען
    msg['Subject'] = נושא

    # הוספת תוכן המייל
    msg.attach(MIMEText(תוכן, 'plain', 'utf-8'))

    # צירוף הקובץ
    if os.path.exists(נתיב_קובץ):
        שם_קובץ = os.path.basename(נתיב_קובץ)

        # קריאת הקובץ
        with open(נתיב_קובץ, 'rb') as קובץ:
            חלק = MIMEBase('application', 'octet-stream')
            חלק.set_payload(קובץ.read())

        # קידוד הקובץ לטקסט (base64) כדי שיעבור בבטחה במייל
        encoders.encode_base64(חלק)

        # שם קובץ בעברית חייב קידוד מיוחד: filename=('utf-8', '', שם)
        חלק.add_header(
            'Content-Disposition',
            'attachment',
            filename=('utf-8', '', שם_קובץ)
        )

        msg.attach(חלק)
        print(f"הקובץ '{שם_קובץ}' צורף בהצלחה!")
    else:
        print(f"שגיאה: הקובץ {נתיב_קובץ} לא נמצא")

    return msg


# דוגמה לשימוש
msg = בנה_מייל_עם_קובץ(
    נמען='boss@company.com',
    נושא='דוח חודשי - יוני 2024',
    תוכן='שלום,\n\nמצורף דוח המכירות החודשי.\n\nבברכה',
    נתיב_קובץ='דוח_יוני.xlsx'
)

# מה כתוב ב-header של הקובץ המצורף?
for חלק in msg.get_payload():
    if חלק.get_filename():
        print("Content-Disposition:", חלק['Content-Disposition'])
        print("שם הקובץ אחרי פענוח:", חלק.get_filename())

# שומרים לקובץ eml כדי לראות את המייל בתוכנת המייל
with open('report_june.eml', 'wb') as f:
    f.write(msg.as_bytes())
print("\nהמייל עם הקובץ המצורף נשמר ב-report_june.eml")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
הפונקציה <strong>בונה</strong> מייל עם קובץ מצורף (ולכן קראנו לה <code>בנה_מייל_עם_קובץ</code> ולא "שלח" - היא לא שולחת כלום).<br>
תחילה יצרנו בעצמנו קובץ Excel כדי שתמיד יהיה מה לצרף. אחר כך הקובץ נקרא כבייטים, קודד ל-base64 והוסף כחלק נוסף במייל.<br>
שם קובץ בעברית דורש את הצורה <code>filename=('utf-8', '', שם)</code>. אם נכתוב סתם <code>filename=שם</code>, חלק מתוכנות המייל יציגו שם משובש.<br>
בסוף שמרנו את המייל ל-<code>report_june.eml</code>: פתחו אותו וראו את הקובץ המצורף.
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">שליחה לכמה נמענים</p>

לפעמים צריך לשלוח את אותו מייל לרשימת אנשים.
</p>

```python
def בנה_מייל_לרשימה(רשימת_נמענים, נושא, תוכן):
    """
    בונה שתי גרסאות של מייל לרשימת נמענים (לא שולח!)

    Args:
        רשימת_נמענים: רשימה של כתובות מייל
        נושא: נושא המייל
        תוכן: תוכן המייל
    """
    # דרך 1: To - כולם רואים את כל הכתובות
    msg = MIMEMultipart()
    msg['From'] = 'sender@example.com'
    msg['To'] = ', '.join(רשימת_נמענים)  # מחבר עם פסיקים
    msg['Subject'] = נושא
    msg.attach(MIMEText(תוכן, 'plain', 'utf-8'))

    # דרך 2: Bcc - אף אחד לא רואה את האחרים
    # בשדה To שמים את השולח עצמו, ואת הנמענים האמיתיים ב-Bcc
    msg2 = MIMEMultipart()
    msg2['From'] = 'sender@example.com'
    msg2['To'] = 'sender@example.com'
    msg2['Bcc'] = ', '.join(רשימת_נמענים)
    msg2['Subject'] = נושא
    msg2.attach(MIMEText(תוכן, 'plain', 'utf-8'))

    return msg, msg2


# דוגמה
נמענים = [
    'employee1@company.com',
    'employee2@company.com',
    'employee3@company.com'
]

msg, msg_bcc = בנה_מייל_לרשימה(
    נמענים,
    'עדכון חשוב מהנהלה',
    'שלום לכולם,\n\nרצינו לעדכן...'
)

print(f"גרסה 1 (To) - הכותרת To: {msg['To']}")
print(f"גרסה 2 (Bcc) - הכותרת To: {msg_bcc['To']}")
print(f"גרסה 2 (Bcc) - הכותרת Bcc: {msg_bcc['Bcc']}")
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מתי להשתמש במה?</strong><br>
<code>To</code> - כשכולם צריכים לדעת מי עוד קיבל את המייל (דיון קבוצתי).<br>
<code>Cc</code> - "לידיעה" - אנשים שצריכים לדעת אבל לא חייבים לענות.<br>
<code>Bcc</code> - כשכל אחד צריך לקבל אבל לא לדעת מי עוד (ניוזלטרים, התראות).
</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>איך Bcc עובד באמת?</strong><br>
כששולחים עם <code>server.send_message(msg)</code>, הפונקציה אוספת את הנמענים מהשדות <code>To</code>, <code>Cc</code> ו-<code>Bcc</code>,
ושולחת לכולם. את השדה <code>Bcc</code> עצמו היא <strong>מסירה</strong> מהעותק שנשלח - ולכן אף נמען לא רואה את הרשימה.<br>
זו הסיבה שב-Bcc לא מכניסים את הרשימה לשדה <code>To</code>.
</p>

## <p style="text-align: right; direction: rtl; float: right; clear: both;">מחלקה מלאה לשליחת מיילים</p>

<p style="text-align: right; direction: rtl; float: right; clear: both;">
הנה מחלקה מקיפה שמשלבת את הכל.<br>
שימו לב לחלוקה: <code>build_message</code> רק <strong>בונה</strong> את המייל (בטוח להריץ בכיתה),
ו-<code>send_email</code> בונה <strong>ושולח</strong> באמת (רק בבית).
</p>

```python
import smtplib
from email.mime.text import MIMEText
from email.mime.multipart import MIMEMultipart
from email.mime.base import MIMEBase
from email import encoders
import os


class EmailSender:
    """מחלקה לשליחת מיילים בצורה נוחה"""

    def __init__(self, smtp_server, port, email, password):
        self.smtp_server = smtp_server
        self.port = port
        self.email = email
        self.password = password

    def build_message(self, to, subject, body, attachments=None, html=False):
        """
        בניית מייל (בלי לשלוח!)

        Args:
            to: נמען או רשימת נמענים
            subject: נושא
            body: תוכן המייל
            attachments: רשימת נתיבי קבצים לצירוף (אופציונלי)
            html: האם התוכן הוא HTML (False = טקסט רגיל)
        """
        msg = MIMEMultipart()
        msg['From'] = self.email
        msg['To'] = to if isinstance(to, str) else ', '.join(to)
        msg['Subject'] = subject

        # הוספת תוכן
        content_type = 'html' if html else 'plain'
        msg.attach(MIMEText(body, content_type, 'utf-8'))

        # צירוף קבצים
        if attachments:
            for file_path in attachments:
                if os.path.exists(file_path):
                    self._attach_file(msg, file_path)
        return msg

    def send_email(self, to, subject, body, attachments=None, html=False):
        """בונה ושולח מייל באמת - להרצה בבית בלבד"""
        msg = self.build_message(to, subject, body, attachments, html)
        try:
            server = smtplib.SMTP(self.smtp_server, self.port, timeout=10)
            server.starttls()
            server.login(self.email, self.password)
            server.send_message(msg)
            server.quit()
            return True, "המייל נשלח בהצלחה!"
        except Exception as e:
            return False, f"שגיאה בשליחה: {e}"

    def _attach_file(self, msg, file_path):
        """צירוף קובץ למייל"""
        filename = os.path.basename(file_path)
        with open(file_path, 'rb') as f:
            part = MIMEBase('application', 'octet-stream')
            part.set_payload(f.read())
        encoders.encode_base64(part)
        part.add_header('Content-Disposition', 'attachment', filename=('utf-8', '', filename))
        msg.attach(part)


# בכיתה: בונים מייל עם פרטים לא אמיתיים ושומרים אותו לקובץ
sender = EmailSender(
    smtp_server='smtp.gmail.com',
    port=587,
    email='your-email@gmail.com',
    password='your-app-password'
)

msg = sender.build_message(
    to='recipient@example.com',
    subject='בדיקה',
    body='זהו מייל בדיקה',
    attachments=['דוח_יוני.xlsx']
)
with open('class_mail.eml', 'wb') as f:
    f.write(msg.as_bytes())
print("המחלקה EmailSender מוכנה! נבנה מייל ונשמר ב-class_mail.eml")

# שליחה אמיתית - רק בבית, אחרי שמילאתם פרטים אמיתיים:
# הצלחה, הודעה = sender.send_email(
#     to='recipient@example.com',
#     subject='בדיקה',
#     body='זהו מייל בדיקה',
#     attachments=['דוח_יוני.xlsx']
# )
# print(הצלחה, הודעה)
```

<p style="text-align: right; direction: rtl; float: right; clear: both;">
<strong>מה קרה כאן?</strong><br>
חילקנו את העבודה לשני חלקים: <code>build_message</code> בונה מייל (אפשר לבדוק אותו, להדפיס ולשמור ל-<code>.eml</code>),
ו-<code>send_email</code> מוסיף את ההתחברות והשליחה.<br>
חלוקה כזו נפוצה בקוד אמיתי: כך אפשר לבדוק את החלק הראשון בלי שרת מייל.
